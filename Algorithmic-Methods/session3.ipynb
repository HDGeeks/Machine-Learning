**2.3 Dictionaries**

A dictionary is another data structure that functions as storage for objects.
Entries in dictionaries are realised as **key - value** pairs.

Instead of referencing an entry by index (as with lists), they are referenced by their **key**.

Dictionaries are intialised by using curly brackets {}. Entries are separated by commas and consist of **KEY : VALUE**

**ATTENTION:** Only immutable data types can be used as keys!

```python
dictionaryDE_RU = {'Feige' : 'Инжир', 'Auto' : 'Машина'}
dictionaryDE_RU
```

```python
dictionaryDE_RU['Feige'] #referencing entries by key
```

Values can be manipulated and or replaced

```python
newDictionary = {'a' : 1, 'b' : 2, 'c' : 3}
newDictionary['a'] += 10 #change the value for key 'a'
newDictionary
```

```python
del newDictionary['a'] #remove the key 'a' (and its corresponding value!)
newDictionary
```

Note that keys in a dictionary are always unique! If you try to insert an existing key, it will overwrite the old entry:

```python
newDictionary['b'] = 10 #overwrite the old key/value pair
newDictionary
```

To get a list of all keys or values, the *keys()* and *values()* methods can be used:

```python
newDictionary.keys() #return all keys
```

```python
newDictionary.values() #return all values
```

```python
type(newDictionary.values()) #note that technically speaking, the return type is not a list!
```

Keys can only be immutable data types! Values can be of any type; even lists and other dictionaries:

```python
errorDict = {} #to demonstrate what will happen if we set a mutable data type as key
listKey = [1]
errorDict[listKey] = 1 #this will not work
```

```python
listDict = {60:[4,3,2,1],124:[6,3,6,1]}
listDict[124]
```

```python
dictOfDicts = {} #dictionaries within a dictionary
dictOfDicts[1] = dictionaryDE_RU
dictOfDicts[2] = listDict
dictOfDicts
```

```python
dictOfDicts[1]['Auto'] #access an element in a dictionary of dictionaries by providing multiple keys
```

There are no further restrictions (with the exception of keys being immutable) for combinations of data types within dictionaries:

```python
mixedDict = {}
mixedDict[1] = "Hi"
mixedDict['another key'] = 123
mixedDict[5.1] = mixedDict #Possible, but does not provide any benefits!

mixedDict[5.1]
```

**2.4. Comparison - List vs. Dictionary**

Both can store objects.

For both, there is no restriction regarding the data types (or combination of data types) they can contain

List elements are referenced by their index:

    Elements: 'Hello', 'my', 'friends'
    Indices:   0        1     2


Dictionaries consist of key/value pairs. A value is referenced by its key:

    Keys:   Car   House   Street
    Values: Auto  Haus    Straße

Lists are *ordered* collections of elements, whereas Dictionaries are *unordered*. This means that for lists, the order in which elements are added is important:

```python
a = []
a.append(1)
a.append(2)
a.append(3)
a
```

```python
b = []
b.append(1)
b.append(3)
b.append(2)
b
```

```python
c = []
c.append(1)
c.append(2)
c.append(3)

c==a #check if lists are the same (i.e. same elements and same order)
```

```python
a == b
```

For dictionaries, order is not relevant. If they contain the same key/value pairs, they are considered to be equal:

```python
a = {}
a[1] = 1
a[2] = 2
a[3] = 3
a
```

```python
b = {}
b[1] = 1
b[3] = 3 #different order of creating key/value pairs!
b[2] = 2
b
```

```python
b == a
```

```python
c = {}
c[1] = 1
c[2] = 2
c[3] = 3
c == a
```

**2.5. Boolean values**

In the example above, the equality operator *==* was introduced.

The result of such an equality comparison is a boolean value. A boolean value is a special numerical data type that can only take up the values 0 or 1.
It can be thought of as a 1-bit Integer. To distinguish between boolean values and Integers, the zero is expressed by the keyword **False**, and the one is expressed by the keyword **True** (Note that the first letter is capitalized!):


```python
is_asleep = True
is_asleep
```

```python
int(is_asleep) #convert to integer
```

```python
is_hungry = False
is_hungry
```

```python
int(is_hungry)
```

```python
is_asleep = 1
bool(is_asleep) #convert back to boolean
```

```python
is_asleep = 0
bool(is_asleep)
```

Note that if converted to bool, every Integer that is not Zero will equal to TRUE:

```python
is_asleep = -1
bool(is_asleep)
```

```python
is_asleep = 91919
bool(is_asleep)
```

Note that there are two different concepts of identity:

Identity of values (or list contents, etc.), expressed by the **==** operator:

```python
a = [1,2,3]
b = [1,2,3]
a == b
```

Identity of objects, expressed by the id() function.

```python
id(a) == id(b)
```

Immutable data types (like strings, integers, floats, etc.) will also count as identical objects, even if they are inside different variables:

```python
a = "hello"
b = "hello"
a == b
```

```python
id(a) == id(b)
```

Booleans are also returned when you check whether a value is contained in a list by using the **in** operator:

```python
newList = [515,12,56]
12 in newList #checks is 12 is contained in the list
```

```python
919 in newList
```

Comparisons like greater than, equal to and lesser than (Operators: >,>=,==,<,<=) also return boolean values:

```python
1 >= 1
```

```python
50 < 51
```

```python
0 > 9
```

**2.7. Data Type Summary**

[Objects](https://docs.python.org/3/reference/datamodel.html) are Python’s abstraction for data. All data in a Python program is represented by objects or by relations between objects. [...]

**Every object has an identity, a type and a value**. An object’s identity never changes once it has been created; you may think of it as the object’s address in memory. The is operator compares the identity of two objects; the id() function returns an integer representing its identity.

**An object’s type determines the operations that the object supports** (e.g., “does it have a length?”) and also defines the possible values for objects of that type. The type() function returns an object’s type (which is an object itself). Like its identity, an object’s type is also unchangeable

```python
integerA = 5 #integer - immutable
```

```python
id(integerA) #identity
```

```python
type(integerA) #data type
```

```python
integerA #value
```

```python
listA = [1,2,3] #list - mutable
```

```python
id(listA)
```

```python
listA.append(4)
listA #value has changed
```

```python
id(listA) #identity is the same as before
```

**The value of some objects can change**. Objects whose value can change are said to be mutable; objects whose value is unchangeable once they are created are called immutable. (The value of an immutable container object that contains a reference to a mutable object can change when the latter’s value is changed; however the container is still considered immutable, because the collection of objects it contains cannot be changed. So, immutability is not strictly the same as having an unchangeable value, it is more subtle.) An object’s mutability is determined by its type; for instance, numbers, strings and tuples are immutable, while dictionaries and lists are mutable.

**Some objects contain references to other objects**; these are called containers. Examples of containers are tuples, lists and dictionaries. The references are part of a container’s value. In most cases, when we talk about the value of a container, we imply the values, not the identities of the contained objects; however, when we talk about the mutability of a container, only the identities of the immediately contained objects are implied. So, if an immutable container (like a tuple) contains a reference to a mutable object, its value changes if that mutable object is changed.

```python
list1 = [1,2,3] #list, mutable
list2 = [4,5,6]
tupleA = (list1,list2) #immutable tuple containing two lists
tupleA
```

```python
list1.append(10)
tupleA #immutability refers to the identity of the contained elements, therefore, the value can change, as long as the elements remain the same!
```

**split()**
>  str.split(sep=None, maxsplit=-1)

> Return a list of the words in the string, using sep as the delimiter string. If maxsplit is given, at most maxsplit splits are done (thus, the list will have at most maxsplit+1 elements). If maxsplit is not specified or -1, then there is no limit on the number of splits (all possible splits are made).

> If sep is given, consecutive delimiters are not grouped together and are deemed to delimit empty strings (for example, '1,,2'.split(',') returns ['1', '', '2']). The sep argument may consist of multiple characters as a single delimiter (to split with multiple delimiters, use re.split()). Splitting an empty string with a specified separator returns [''].



```python
'1,2,3'.split(',')
```

```python
type('1,2,3'.split(','))
```

```python
'this is a test'.split(' ')
```

> If sep is not specified or is None, a different splitting algorithm is applied: runs of consecutive whitespace are regarded as a single separator, and the result will contain no empty strings at the start or end if the string has leading or trailing whitespace. Consequently, splitting an empty string or a string consisting of just whitespace with a None separator returns [].

```python
'this is a test'.split()
```

```python
' this   is a test   '.split(' ')
```

```python
' this   is a test   '.split()
```

**2.9. Example Application: Text processing**

DISCLAIMER: The methods and procedures shown here are not necessarily the best to solve the given task. The focus is on showing examples and introducing some new methods.

```python
randomWikipediaArticle = "Risby is the site of a deserted village and former stately home in the civil parish of Rowley in the East Riding of Yorkshire, England. \
It is approximately 3.5 miles (5.6 km) south-west of Beverley and 1 mile (1.6 km) west of the A164 road. The area has an open partially wooded parkland, \
once one of the largest deer parks in Yorkshire, that is popular with local walkers. It also includes Risby Park, a farm, the Folly Lake Cafe. \
There are several fishponds in the ornamental lakes of the former hall, which have been operated as coarse fishing locations since 1990. \
The site also has an octagonal brick folly that is designated Grade II."
randomWikipediaArticle
```

At first we want to know how many words there are in the article:

```python
randomWikipediaArticle.count(' ') #our first approach is to count whitespaces
```

The example above provides a rough estimate. Consider that for an example text with three words "The cat eats", there are only two whitespaces!

A better way would be using the split() method:

```python
wordList = randomWikipediaArticle.split() #split the article into a list of words. The default behavior is to split at whitespaces
wordList
```

```python
len(wordList) #now we can count the number of words
```

In the next step, we want to know how many unique words there are in the text:

```python
wordSet = set(wordList) #convert the list to a set
len(wordSet) #now we can count the number of unique words
```

There are still some downsides to this approach:

```python
wordSet
```

```python
'The' in wordSet
```

```python
'the' in wordSet
```

```python
'wooded' in wordSet
```

```python
'Wooded' in wordSet
```

Word capitalisation matters! A solution for this issue is introduced here:

```python
preprocessedArticle = randomWikipediaArticle.lower() #convert all words to lowercase
preprocessedWordList = preprocessedArticle.split() #split the article into a list of words
preprocessedWordSet = set(preprocessedWordList) #convert the list to a set
len(preprocessedWordSet)
```

Now we count 80 distinct words (as opposed to 82 in the example before)

Other common issues in text processing are handling punctuation (do they belong to a word, a numeral or do they indicate the end of a sentence?) and special characters. In our example, we will ignore the different potential meanings and just remove everything:

```python
preprocessedArticle = preprocessedArticle.replace('.','') #remove a punctuation marks by replacing them with an empty string
preprocessedArticle = preprocessedArticle.replace(',','') #same for commas
preprocessedArticle = preprocessedArticle.replace('"','') #and so on ...
preprocessedArticle = preprocessedArticle.replace('(','')
preprocessedArticle = preprocessedArticle.replace(')','')
preprocessedWordListNew = preprocessedArticle.split()
set(preprocessedWordListNew)
```

Finally, we want to count the number of occurences for each word. To achieve this, we have to go over all elements in the text (in programming terms, we say: *iterate over the text*), save yet unseen elements and keep count of their occurences. This can be realised by using a *loop*:

```python
countDictionary = {} #init empty dictionary; this will be used to store the words and their corresponding occurences
for word in preprocessedWordListNew: #iterate over the wordlist
  if word not in countDictionary: #if the current word is previously unseen, we create a new entry in the dictionary with the word as key and the number 1 as value
    countDictionary[word] = 1
  else: #if an entry with the current word as key already exists, we increase the value by 1
    countDictionary[word] += 1
countDictionary
```
