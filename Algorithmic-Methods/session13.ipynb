# **Plotting with Matplotlib (Overview)**
Matplotlib provides a variety of options for data visualization. For a simple plot of a line in a coordinate system, only  the y and x values are required. Everything else is handled by pyplot. A simple plot can be created by using the *plot()* method and - depending on the current working environment - the *show()* method. (https://matplotlib.org/)

The example above demonstrates a simple line plot of a quadratic function. First, the x and y values are created via list comprehensions:

```python
import matplotlib.pyplot as plt
supported_methods = plt.__dir__()
print(len(supported_methods))
dunder_methods = [method for method in supported_methods if method.startswith("__") and method.endswith("__")]
print(len(dunder_methods))
print(len(supported_methods) - len(dunder_methods))
```

```python
import matplotlib.pyplot as plt

#list comprehension for number -10 to 10
x = [i for i in range(-10,11)]
print(x)
#divide by 5 for smooth display

x = [i/5 for i in x]
print(x)
#square
y = [i**2 for i in x]

print(x)
print(y)
```

In the next step, the *plot()* and *show()* methods are used to create and display the plot. In the most simple case, plot() takes two lists as arguments, one with the x values and one with the y values.

```python
plt.plot(x,y) #plot x and y values
plt.show() #display the plot
```

# Coordinate System
It is possible to add something that resembles a coordinate system by adding a horizontal and a vertical line to the plot:

- *axhline(y)* adds a horizontal line on the y-axis position indicated by the parameter *y*.
- *axvline(x)* adds a vertical line on the x-axis position indicated by the parameter *x*.

In general, adding additional elements to a graph is as simple as calling the corresponding methods and providing the necessary arguments:

```python
#draw cartesian coordinate system
plt.plot(x,y) #plotting the line
plt.axhline(0, color='black') #add a horizontal line at y = 0
plt.axvline(0, color='black') #add a vertical line at x = 0
plt.show()
```

# Adding a grid
A grid can improve the readability of a figure by adding lines to the background.

The most basic grid can be added by using the *grid()* method:

```python
#add grid
plt.plot(x,y)
plt.axhline(y=0, color='black')
plt.axvline(x=0, color='black')
plt.grid() #add a grid in the background for better overview
plt.show()
```

It is also possible to display only vertical or horizontal lines by using the *axis* parameter. For many plot elements, an option for different coloring is also available:

**NOTE:** In general, colors should not be used as excessively as demonstrated on the figure below!

```python
#add grid
plt.plot(x,y)
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid(axis='y',color='red') #add a vertical grid
plt.grid(axis='x',color='green') #add a horizontal grid
plt.show()
```

# Adding labels
By adding labels to the individual axis, the readability of a figure can be improved further. As with all textual elements, different coloring options are available:

```python
#adding labels for the axis
plt.plot(x,y)
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("x-axis") #adding the labels
plt.ylabel("y-axis")
plt.show()
```

# Adding a second line
It is possible to add a second line by providing another set of x,y coordinates and calling the *plot()* method. This can be useful for a direct visual comparison of your data. Note that if not explicitly passed as argument, the second line will default to a different color than the first one:

```python
#adding a linear function
x1 = [i for i in range(-2,3)]
y1 = [2*i+1 for i in x1]

plt.plot(x,y) #plotting the first line
plt.plot(x1,y1) #plotting the second line
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Adding a legend
A legend allows to add a short explanation of the data, indicating the meaning of the different lines in the graph:

```python
#adding a legend
plt.plot(x,y)
plt.plot(x1,y1)
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend(["y = x^2", "y = 2x+1"])
plt.show()
```

# Adding a title

```python
#adding a title
plt.plot(x,y)
plt.plot(x1,y1)
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.legend(["y = x^2", "y = 2x+1"])
plt.title("Some Functions")
plt.show()
```

# Altering individual elements
The position, style and color of some elements can be changed by providing the desired look as parameters in the respective method calls.

One obvious way of changing an element is by changing its color. This can be done by changing the color attribute of an element:

https://matplotlib.org/stable/gallery/color/named_colors.html

```python
plt.plot(x,y,color='green')
plt.show()
```

A marker allows for indicating the exact points provided by the initial data.
An overview of the different marker types can be found here:

https://matplotlib.org/stable/gallery/lines_bars_and_markers/marker_reference.html

```python
plt.plot(x,y,color='green',marker='+')
plt.show()
```

In addition to the markers, a variety of linestyles is also available:

https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html



```python
plt.plot(x,y,color='green',marker='+',linestyle='dashed',linewidth=1,markersize=5)
plt.show()
```

Instead of using the *linestyle*, *marker* and *color* attributes individually, the style can also be defined in a short string like shown below:

    "go--
    g = green
    o = o markers
    -- = dashed line

```python
#Alternative
plt.plot(x,y,'go--',linewidth=1,markersize=5)
plt.show()
```

Label, title and legend positions can also be altered by using the *loc* attribute:

```python
plt.plot(x,y)
plt.plot(x1,y1)
plt.axhline(0, color='black')
plt.axvline(0, color='black')
plt.grid()
plt.xlabel("x-axis",loc='left') #only left, right and center are available
plt.ylabel("y-axis",loc='bottom') #only top, bottom and center are availabe
plt.legend(["y = x^2", "y = 2x+1"])
plt.title("Graph Example",loc='right')
plt.show()
```

- barplots

**Statistical distribution**:
- histogram

# Pairwise data:
All of the following plot types require pair-wise data, like the x and y values from the examples above.

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y) #line plot
plt.show()
```

```python
plt.scatter(x,y) #scatterplot
plt.show()
```

```python
plt.bar(x,y) #barplot
plt.show()
```

```python
#instead of using numerical values for x, strings are also possible in a barplot
x = ["A","B","C","D","E"]
y = [2,4,6,8,10]

plt.bar(x,y)
plt.show()
```

# Statistical distributions
These graphs show how a variable is distributed. A common example is a frequency distribution, which shows how many instances of some variable x occur once, twice, etc.

An example would be the age of participants in a university class:

```python
#some age distribution:
ages = [18,19,20,21,22,23,24,25,26,29,20,20,20,21,39]

plt.hist(ages)
plt.show()
```

It can be seen from the code example that the actual distribution is calculated by the *hist()* method. Only the raw data is provided as an argument.

A boxplot can provide a quick overview over some descriptive statistic measure:

- Minimum (Q0 or 0th percentile): the lowest data point in the data set
- Maximum (Q4 or 100th percentile): the highest data point in the data set
- Median (Q2 or 50th percentile): the middle value in the data set
- First quartile (Q1 or 25th percentile): also known as the lower quartile qn(0.25)
- Third quartile (Q3 or 75th percentile): also known as the upper quartile qn(0.75)

```python
#boxplot
plt.boxplot(ages)
plt.grid()
plt.show()
```

```python
plt.ecdf(ages)
plt.grid()
plt.show()
```

# **Example Application**

```python
import requests
import json
from bs4 import BeautifulSoup

page = BeautifulSoup(requests.get("https://www.tagesschau.de/").text, "html.parser") #visit the page

#<a class="teaser__link" -> this is the identified pattern that leads to all articles
articleList = page.find_all("a", class_="teaser__link", href=True)

baseUrl = "https://www.tagesschau.de" #the base url; needs to be combined with the found hrefs
articles = [] #empty list to store articles
for article in articleList:
  if article["href"].startswith("/") and not article["href"].startswith("/multimedia"): #filter out unwanted elements that might cause problems
    articlePage = BeautifulSoup(requests.get(baseUrl + article["href"]).text,"html.parser") #gathering the data
    article = articlePage.find("script", type="application/ld+json").text
    article = json.loads(article)
    articles.append(article)

articles[0].keys()
```

```python
articles[0]['articleBody']
```

```python
!python -m spacy download de_core_news_lg
```

```python
import spacy

nlp = spacy.load("de_core_news_lg")
```

```python
#collect all lemmas in a list
lemmas = []
for article in articles:
  doc = nlp(article['articleBody'])
  for token in doc:
    lemmas.append(token.lemma_)
len(lemmas)
```

```python
import pandas as pd

# Assuming 'lemmas' is the list containing all the extracted lemmas

# 1. Reading lemmas into Pandas DataFrame
df = pd.DataFrame({'lemma': lemmas})

# 2. Determine Their Frequencies
frequency = df['lemma'].value_counts()

# Print the frequencies
print(frequency)
```

By the information above, we can conclude that there are 4473 types and 21698 tokens in the article texts.

```python
#extract the distribution of counts from the df above
counts = frequency.values
counts
```

```python
#display a histogram of the frequency distribution and logarithmix y axis
sns.histplot(counts,bins=200)
plt.yscale('log') #logarithmic y axis
plt.show()
```

```python
#display cumulated sum counts as x axis label
sns.ecdfplot(counts)
plt.xlabel("counts")
plt.xscale('log')
#y axis ticks every 0.1
plt.show()
```
