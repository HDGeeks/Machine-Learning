# IV - Functions (10 points)

Consider the function below for calculating the factorial of a number (Example: 3! = 3 * 2 * 1)

```python
def factorial(number):
  if number == 0:
    return 1
  else:
    return number * factorial(number-1)
  
print(factorial(3))

#Q1 : 1. Outline the individual steps the function above takes to calculate the factorial of 3. (8 points)
# step 1
# input is 3 , since 3 is not 0 , we go to the else part and return 3 * factorial(2)
# step 2
# input is 2 , since 2 is not 0 , we go to the else part and return 2 * factorial(1)
# step 3
# input is 1 , since 1 is not 0 , we go to the else part and return 1 * factorial(0)
# step 4
# input is 0 , since 0 is 0 , we return 1
# step 5
# we go back to step 3 and return 1 * 1 which is 1
# step 6
# we go back to step 2 and return 2 * 1 which is 2
# step 7
# we go back to step 1 and return 3 * 2 which is 6  

# Q2. What is this type of function called? (1 point)
# This type of function is called a recursive function.

# Q3. Can the factorial be calculated in fewer steps using an iterative function? (1 point)
# Yes, the factorial can be calculated in fewer steps using an iterative function,
# since we can use a loop to calculate the factorial without 
# the need for multiple function calls as in the recursive version.

def factorial_iterative(number):
    result = 1
    for i in range(1, number + 1):
        result *= i
    return result
```
