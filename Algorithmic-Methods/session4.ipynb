# **While Loops**

> The while statement is used for repeated execution as long as an expression is true

The *while* keyword is used to construct a while loop. In general terms, the syntax works as follows:

    while *condition*:
      *execute statement*

This means that a condition in a while loop must be a boolean value. If the condition is of a different type, the interpreter tries to convert it to a boolean value:

```python
display(bool([])) #False
display(bool([1])) #True
display(bool("Hi")) #True
display(bool("")) #False
```

A while loop can be used to iterate over elements of a list. For this, a *counter* variable is required (commonly named i) and needs to be incremented after each step.

```python
exampleList = [1,2,3,4,5]
i = 0
while i < len(exampleList):
  display(exampleList[i])
  i += 1
```

Be careful to set the correct starting index and ending condition here: **i=0** and **i<len(exampleList)**


In most cases, the condition is designed in such way that the evaluated variable changes at some point, and thus ending the loop. Loops with conditions that are never met are called *Infinite Loops* and will run until the program is interrupted manually:

    while True:
      print ("Still going")

The code snippet above shows a loop that will never terminate because the variable in the condition never changes.


By using only elementary operations and a while loop, we can already express some slightly more advanced mathematical operations like a factorial:

    Factorial:
    5! = 5 * 4 * 3 * 2 * 1
    6! = 6 * 5 * 4 * 3 * 2 * 1
    and so on ...

If we want to express the Factorial of a given number **a**, we can make use of a while loop:

```python
a = 5         #to express a! with a = 5
i = a-1       #counter variable that starts at a-1
while i > 0:  #we count backwards until i equals 1
  a = a * i   #overwrite the initial value with its intermediate value
  i -= 1      #substract 1 from the counter variable
display(a)
```

```python
a = 5
i = a-1
while i > 0:
  print(f"Calculation:{a} * {i}")
  a = a * i
  i -= 1
  print(f"Values after Iteration: a = {a}, i = {i}\n")
print (f"Final Result: {a}")
```

Another application would be to let the program running until a certain input is detected:

```python
password = "1234"

passwordEntered = False

while not passwordEntered:
  current = input("Enter password:")
  if current == password:
    passwordEntered = True
  else:
    print("Wrong password!")

print ("Login successful")
```

# **For Loops**

> The for statement is used to iterate over the elements of a sequence (such as a string, tuple or list) or other iterable object



Instead of using the while loop from the example above, iterating over a list can be done with a for loop:

```python
for element in exampleList:
  display(element)
```

The for loop can be used for all iterable data types (like tuples, lists, dict, set)

```python
exampleSet = set([1,2,3,4,5])
for element in exampleSet:
  display(element)
```

For dictionaries, the syntax above only prints the keys:

```python
exampleDict = {"a":1, "b":2, "c":3}
for element in exampleDict:
  display(element)
```

In order to get the values, you need to explicity reference the dictionary with the current key, or iterate over the values directly:

```python
for element in exampleDict:
  display(exampleDict[element]) # accessing elements
```

```python
for element in exampleDict.values(): #iterating over values directly
  print (element)
```

Loops can be nested, which is useful for traversing nested datastructures (like lists of lists). Consider the following list of lists:

    example = [[1,2],[5,1],[8,9]]

Each element of this list consists of another list. The proper way to iterate over such a data structure would be a nested for loop.

The first loop iterates over the outer list, while the second loop iterates over the elements inside the inner lists:

```python
example = [[1,2],[5,1],[8,9]]
for lists in example:
  display(lists) #Outer Loop
  for element in lists:
    display(element) #Inner Loop
```

If you want to display both indices and values while iterating over a list, the *range()* function* can be used. It returns an arithmetic progression (for instance the numbers 1 to 10), which then can be used to iterate over a list:

```python
for i in range(len(example)):
  print(i, example[i])
  for j in range(len(example[i])): #j because i is already in use!
    print(f"Index:({i,j})\tValue: {example[i][j]}")
```

**[range()](https://docs.python.org/3/library/stdtypes.html#range)**



> The range type represents an immutable sequence of numbers and is commonly used for looping a specific number of times in for loops.

...

> The advantage of the range type over a regular list or tuple is that a range object will always take the same (small) amount of memory, no matter the size of the range it represents (as it only stores the start, stop and step values, calculating individual items and subranges as needed).

```python
range(10)
```

This is sufficient for iteration purposes, as the values themselves will be produced on the fly:

```python
for number in range(10):
  print (number)
```

Checks if a certain element is contained in a range are performed instantaneously:

```python
10 in range(10000)
```

```python
999999 in range(10000)
```

The *step* parameter can be set to skip elements between start and stop:

```python
for number in range(0,10,3):
  print (number)
```

range() objects can be transformed to lists if required:

```python
list(range(10)) #quickly fills a list with numbers zero to 9
```

# **Checking conditions (if, elif, else)**
Before each iteration, the while loop checks if a condition is still met. The while loop executes its instructions as long as the condition is not met anymore.

The *if* statement also allows to check if a condition is met, but will only execute the instructions within the statement once. When the behavior of a program should change based on certain conditions, like the current state of a variable, *if - else* blocks can be used for control flow.

    if *condition* :
      *ececute statement*
    else:
      *execute alternative statement*

If multiple conditions need to be checked, the *elif* keyword can be used:

    if *condition* :
      *execute statement*
    elif *other condition* :
      *execute option 2*
    else:
      *execute alternative statement*

The statements within an if, or elif clause are only executed if the condition is met. The statement within an else clause are executed if none of the former conditions are met.

```python
animal = "cat"
if animal == "cat":
  print("Meow")
else:
  print("Not a cat!")
```

If multiple if statements (instead of if and elif) are used, the second if statement will also be evaluated and (if the condition is met) executed:

```python
animal = "cat"
if animal == "cat": #first is met
  print("Meow") #code is executed
if "a" in animal: #second is also met
  print("Maybe a cat") #code is executed
else:
  print("Not a cat!")
```

```python
animal = "cat"
if animal == "cat": #condition is met
  print("Meow") #first statement will be executed
elif "a" in animal: #second will only be checked if first condition is not met!
  print("Maybe a cat")
else:
  print("Not a cat!")
```

If you work with boolean variables, it is not necessary to explicity check for the value of the variable:

```python
students_asleep = True
if students_asleep: #it is not necessary to check *if students_asleep == True*, because students_asleep already is a boolean value
  print("Students are asleep")
else:
  print("Students are awake")
```

# **Combining Conditions (and, or, not)**
Conditions can be combined by using the logical *and*, *or* and *not* operators to implement basic logic operations

$x_1$|$AND$|$x_2$|=|$x_3$
---|---|---|---|---
False||False||True
False||True||False
True||False||False
True||True||True

```python
#for some reasons, the following conditions must be met for a username: First letter must be uppercase, the number 2 must be included, the last letter must be lowercase and the letter z must not be used

userName = input("Enter your Name: ")
if userName[0].isupper() and "2" in userName and userName[-1].islower() and "z" not in userName:
  print("Username accepted")
else:
  print("Username not accepted")
```

$x_1$|$OR$|$x_2$|$=$|$x_3$
---|---|---|---|---
False||False||False
False||True||True
True||False||True
True||True||True

```python
userRole = "Admin"

if userRole == "Admin" or userRole == "Moderator": #check if the first, the second, or both conditions are met
  print("Access granted")
else:
  print("Access denied")
```

$NOT$|$x_1$|$=$|$x_2$
---|---|---|---
$ $|False||True
$ $|True||False

```python
countDict = {}
listToCount = ["a","a","b","b","b","c","d"]
for element in listToCount:
  if element not in countDict.keys(): #if the key does not exist, we create it
    countDict[element] = 1
  else: #otherwise if will be incremented
    countDict[element] += 1

print (countDict)
```

The same functionality can often be expressed both as negated and non-negated version:

```python
countDict = {}
listToCount = ["a","a","b","b","b","c","d"]
for element in listToCount:
  if element in countDict.keys(): #if the element exists, we increment it
    countDict[element] += 1
  else: #otherwise it will be created
    countDict[element] = 1

print (countDict)
```

# **Caution:**

**Don't manipulate the object you are currently iterating through!**

```python
#The task here is to clean a list of strings by removing all entries that are not email adresses!

textList = ["abc@mail.com","pu@la.com","aw@uni-trier.de","hallo123","asd"] #intitial list

for email in textList: #iterate over all elements
  if "@" not in email: #if @ is not inside the string, it cannot be an email adress
    textList.remove(email) #remove element from list!
display(textList) #print final list
```

Instead, use intermediary variables or introduce new variables to save the entries that you want to keep:

```python
textList = ["abc@mail.com","pu@la.com","aw@uni-trier.de","hallo123","asd"]
keepList = []

for email in textList: #iterate over all elements
  if "@" in email: #if @ is inside the string, it is an email adres
    keepList.append(email) #add element to new list

textList = keepList #overwrite old list
display(textList) #print final list
```
