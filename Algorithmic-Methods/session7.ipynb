# **Recursion**

In some situations it can be beneficial for a function to call itself. This can be used in some sorting and search functions (among many other usecases). Recursion is an important concept in programming languages and can lead to elegant (albeit often inefficient) solutions. Consider the factorial calculation one of the previous sessions:


**Recap - Factorial**

By using only elementary operations and a while loop, we can already express some slightly more advanced mathematical operations like a factorial:

    Factorial:
    5! = 5 * 4 * 3 * 2 * 1
    6! = 6 * 5 * 4 * 3 * 2 * 1
    and so on ...
    Exception: 0! = 1

If we want to express the Factorial of a given number **number**, we can make use of a while loop:

```python
def fac(number):
  if number == 0:           #return 1 if number is 0
    return 1
  else:
    i = number-1            #counter variable that starts at a-1
    while i > 0:            #we count backwards until i equals 1
      number = number * i   #overwrite the initial value with its intermediate value
      i -= 1                #substract 1 from the counter variabl
    return number

fac(5)
```

A more general definition of the factorial could look like this:
    Factorial:
    n! = n * (n-1)!
    0! = 1

    Example:
    4! = 4 * 3 * 2 * 1 => 24
    5! = 5 * 4! => 5 * 24 => 120
  
The definition *n! = n * (n-1)!* is the general case, and the definition of *0! = 1* is the base case.

With both, a recursive function for the calculation of a factorial is possible:

```python
def facRec(number):
  if number == 0:
    return 1
  else:
    return number*facRec(number-1)

facRec(5)
```

# **Listing elements of a nested data structure**
Another example application for recursive algorithms is the traversal of tree-like datastructures. This could be nested lists, directories in an operating system, or html pages.

Consider the following nested list:

```python
nestedList = [5,4,3,[5,1,2,[59,1,25,[5]]],[5,[519,218,[9,5,[515]]]]]
```

For tasks like determining the number of individual elements in a nested list, or for calculating the total sum of all elements in the list, it is not possible to simply use the sum() or len() functions:

```python
sum(nestedList) #not a supported operation
```

```python
len(nestedList) #only lists the elements of the outer list
```

Since the depth of the nesting is not necessarily known, using nested for loops is also not a viable option. Instead, we iterate over the outer list and recursively

```python
def sumElements(someList):
  print (someList)
  nr = 0
  for element in someList:
    if type(element) == type([]): #if the element is another list
      nr += sumElements(element) #recursive call
    else: #otherwise
      nr += element #just adding the element
  return nr

sumElements(nestedList)
```

# **Testing your code (Part 1)**

To verify the functionality of newly implemented functions methods and classes, is useful to deploy basic testing methods.

Basic tests should verify the program behavior for unexpected/unintended input (negative values, non-fitting data types), for edge cases (like empty lists, maximum/minimum numbers) and in general should verify the correct execution of every control flow condition.

Take the following Linked List as an example:


```python
class ListElement():
  def __init__(self,value):
    self.value = value
    self.next = None

  def __str__(self):
    return str(self.value)

class LinkedList():
  def __init__(self):
    self.first = None

  def __str__(self):
    current = self.first
    strRep = "["
    while current != None:
      strRep += f" {current} "
      current = current.next
    strRep += "]"
    return strRep

  def addElementAtPosition(self,value,position):
    newElement = ListElement(value)
    if self.first == None and position == 0: #if list is empty and position is 0
      self.first = newElement
    elif position == 0: #if list is not empty and position is 0
      newElement.next = self.first
      self.first = newElement
    else: #for all other cases
      currentPos = 0
      current = self.first
      while not current.next == None and currentPos < position-1: #if position > length, element will be added at the end of the list
        current = current.next
        currentPos += 1
      newElement.next = current.next
      current.next = newElement
```

To verify the intended behavior of the implementation, the different possible inputs and states should be checked:

- initializing an empty list
- adding an element to an empty list
  - at position 0
  - at position > 0
- adding an element to a non-empty list
  - at position 0
  - at position > 0
  - in-between existing elements (for a list with more than 1 element)
- adding elements of various data types (including None-type)
- testing the string method for empty/non-empty list

```python
#testing for empty list
newList = LinkedList() #init
print (newList)
newList.addElementAtPosition(5,0) #add element at 0

print (newList)
```

```python
#testing for empty list
newList = LinkedList() #init
newList.addElementAtPosition(5,1) #add element at pos>0

print (newList)
```

Evidently, the test-case above is not covered by the implementation!

The following line checks only for emptyness when the second condition is also met:

`if self.first == None and position == 0: `

This can be easily fixed by ignoring the position argument for an empty list, by changing the line to:

`if self.first == None`

```python
class ListElement():
  def __init__(self,value):
    self.value = value
    self.next = None

  def __str__(self):
    return str(self.value)

class LinkedList():
  def __init__(self):
    self.first = None

  def __str__(self):
    current = self.first
    strRep = "["
    while current != None:
      strRep += f" {current} "
      current = current.next
    strRep += "]"
    return strRep

  def addElementAtPosition(self,value,position):
    newElement = ListElement(value)
    if self.first == None: #if list is empty (ignoring position)
      self.first = newElement
    elif position == 0: #if list is not empty and position is 0
      newElement.next = self.first
      self.first = newElement
    else: #for all other cases
      currentPos = 0
      current = self.first
      while not current.next == None and currentPos < position-1: #if position > length, element will be added at the end of the list
        current = current.next
        currentPos += 1
      newElement.next = current.next
      current.next = newElement
```
