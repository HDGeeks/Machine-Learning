## Block 1: Wetterstation - Arbeiten mit Listen 🌡️

### Datensatz: Temperaturmessungen

```python
# Stündliche Temperaturmessungen über einen Tag (24 Werte)
temperatures = [12.5, 13.0, 11.5, 11.0, 10.5, 10.0, 11.5, 14.0, 
                16.5, 18.0, 19.5, 21.0, 22.5, 23.0, 22.5, 21.0,
                20.0, 18.5, 17.0, 15.5, 14.0, 13.5, 13.0, 12.5]

# Liste mit den Uhrzeiten für jede Messung
measurement_times= []
for i in range(24):
    measurement_times.append(f"{i:02d}:00")

# i:02d sorgt dafür, dass die Zahl immer zweistellig ist (z.B. 01, 02, ..., 23)

print("Temperaturen:", temperatures)
print("Messzeiten:", measurement_times)

#measurement_times = [f"{i:02d}:00" for i in range(24)]
```

### Demo: Grundlegende Analysen


```python
# 1. Durchschnittstemperatur berechnen
# Die Funktion sum() addiert alle Werte in der Liste
# Die Funktion len() gibt die Anzahl der Elemente in der Liste zurück
average_temp = sum(temperatures) / len(temperatures)
print(f"Durchschnittstemperatur: {average_temp:.1f}°C")
```

### Aufgaben:

#### 1. Maximum und Minimum finden


```python
# 🎯 Aufgabe 2.1 
# Finden Sie die Tageshöchsttemperatur und die niedrigste Temperatur
# Nutzen Sie dazu die offizielle Python Dokumentation: https://docs.python.org/3/

max_temp = max(temperatures)# Hier Ihre Lösung einfügen 
min_temp = min (temperatures)# Hier Ihre Lösung einfügen

print(f"Höchsttemperatur: {max_temp}°C")
print(f"Tiefsttemperatur: {min_temp}°C")
```

#### 2. Tageszeiten analysieren


```python
# Aufgabe 3.1 
# Teilen Sie die Tagestemperaturen in Tag und Nacht auf
morning_temps = temperatures[6:18]# 6:00 - 18:00 Uhr
night_temps = temperatures[:6] + temperatures[18:]# 18:00 - 06:00 Uhr

# Berechnen Sie die Durchschnittstemperatur für Tag und Nacht
# Verwenden Sie dazu die bekannten Funktionen (https://docs.python.org/3/)
durschnitt_morgen = sum(morning_temps)/len(morning_temps)
durschnitt_nacht = sum(night_temps)/len(night_temps)

print(f"Durchschnittstemperatur Tag: {durschnitt_morgen}°C")
print(f"Durchschnittstemperatur Nacht: {durschnitt_nacht}°C")
```

## Block 3: Warenkorb - Arbeiten mit Dictionaries

### Beispiel-Dictionary:

```python
warenkorb = {
    "Laptop": 800,
    "Maus": 25,
    "Tastatur": 50
}
```

### Grundlegende Operationen:


```python
# Ausgabe des kompletten Warenkorbs
print("Warenkorb:", warenkorb)
```

```python
# Zugriff auf einzelne Werte
laptop_preis = warenkorb["Laptop"]
print("Laptop kostet:", laptop_preis, "€")
```

```python
# Neues Produkt hinzufügen
warenkorb["Bildschirm"] = 200
warenkorb
```

```python
# Preis ändern
warenkorb["Maus"] = 30
warenkorb
```

```python
# Produkt entfernen
del warenkorb["Tastatur"]
warenkorb
```

```python
# Prüfen ob Produkt vorhanden ist
hat_laptop = "Laptop" in warenkorb
hat_laptop
```
