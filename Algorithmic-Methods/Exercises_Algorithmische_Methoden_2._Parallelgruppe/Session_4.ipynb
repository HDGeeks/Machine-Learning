### Exercise 2: String-Manipulation


```python
text = "Python is a nice programming language"

# a) How many characters are inside the string?
character_count = len(text)
print(character_count)
# b) Print out every word in a new line
words = text.split()
for word in words:
    print(word)
# c) Does the text contain the word 'nice'?
if "nice" in text:
    print("Yes, the text contains 'nice'.")
else:
    print("No, the text does not contain 'nice'.")
```

### Exercise 3: Dictionaries


```python
students = {
    "Anna": 19,
    "Carl": 21,
    "Anton": 34,
    "Bernd":20
}


# a) Add a new student

students["Doris"] = 25 # by using the key-value syntax
students.update({"Dora": 25}) # by using the add() method (if it exists, otherwise it will throw an error)

# b) Calculate the average age
average_age = sum(students.values()) / len(students)
print(f"The average age of the students is: {average_age}")
# c) (Nicely) print out every name and corresponding age
for name, age in students.items():
    print(f"{name} is {age} years old.")
```

### Exercise 4: Shopping List

```python
# Create a program for a shopping list:
# 1. A user can insert items via the input() function
# 2. If the user inputs "done", insertion stops
# 3. After insertion, all elements should be printed

items_list = []
item = input("Please enter an item for your shopping list (type 'done' to finish): ")
while item.lower() != "done":
    items_list.append(item)
    item = input("Please enter an item for your shopping list (type 'done' to finish): ")
#
for item in items_list:
    print(item)

# or 
while True:
    item = input("Please enter an item for your shopping list (type 'done' to finish): ")
    if item.lower() == "done":
        break
    items_list.append(item)

print("Your shopping list:")
for item in items_list:
    print(item)
```
