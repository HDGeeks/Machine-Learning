# Exercise V - Regular Expressions

```python
import re
numbers = [1,5,7.5,115.555,.66]
for number in numbers:
    if re.search(r"^\d*\.\d+$",str(number)):
        print (number)
```

- r"..." – Raw-String in Python, damit Backslashes nicht escaped werden müssen
- ^ – Anker für den Anfang des Strings (die Übereinstimmung muss am Anfang beginnen)
- \d* – Null oder mehr Ziffern (0-9) vor dem Dezimalpunkt

-  \d steht für eine Ziffer
-  * bedeutet "null oder mehr Wiederholungen"


- \. – Ein wörtlicher Punkt (Dezimalpunkt) 
Der Backslash escaped den Punkt, da . sonst ein beliebiges Zeichen bedeuten würde


- \d+ – Eine oder mehr Ziffern nach dem Dezimalpunkt

- \d steht für eine Ziffer
- + bedeutet "eine oder mehr Wiederholungen"


- $ – Anker für das Ende des Strings (die Übereinstimmung muss am Ende enden)

```python
def appender(element, list: list):
    list.append(element)
    return list

appender(1, [])
appender(2, [])
appender(3, [])
```
