# Task I:

Use requests and bs4 to collect links to all current articles on the Tagesschau.de news portal. The links should be stored in a list (be sure to save only article links).

Step 1: Visit <a href=https://www.tagesschau.de/>this page</a> and consider how the articles can be found in the HTML source code.

Step 2: Examine the HTML source code and identify the desired elements.

Step 3: Use the find_all() method to retrieve all relevant elements.

```python
import requests
import json #for later use
from bs4 import BeautifulSoup

page = BeautifulSoup(requests.get("https://www.tagesschau.de/").text, "html.parser") #visit the page

#<a class="teaser__link" -> this is the identified pattern that leads to all articles
articleList = page.find_all("a", class_="teaser__link", href=True)

for article in articleList: #iterate over the list and print out the results
    print(article["href"])

# concat the base url with the found href to get the full url of the article
baseUrl = "https://www.tagesschau.de" #the base url; needs to be combined with the found hrefs
articles = [] #empty list to store articles
for article in articleList:
    articleUrl = baseUrl + article["href"] #combine the base url with the found href to get the full url of the article
    articlePage = BeautifulSoup(requests.get(articleUrl).text,"html.parser") #visit the article page
    articleData = json.loads(articlePage.find("script", type="application/ld+json").text) #parse the json data of the article page
    articles.append(articleData) #add the article data to the list of articles
```
