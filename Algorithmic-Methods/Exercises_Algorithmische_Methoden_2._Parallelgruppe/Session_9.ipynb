Verwenden Sie requests und bs4, um Links zu allen aktuellen Artikeln auf dem Nachrichtenportal Tagesschau.de zu sammeln. Die Links sollten in einer Liste gespeichert werden (achten Sie darauf, nur Artikel-Links zu speichern).

Schritt 1: Besuchen Sie <a href=https://www.tagesschau.de/>diese Seite</a> und überlegen Sie, wie sich die Artikel im HTML-Quellcode finden lassen.

Schritt 2: Schauen Sie sich den HTML-Quellcode an und identifizieren Sie die gewünschten Elemente.

Schritt 3: Nutzen Sie die Methode find_all(), um alle relevanten Elemente abzurufen.

```python
import requests
import json #for later use
from bs4 import BeautifulSoup

page = BeautifulSoup(requests.get("https://www.tagesschau.de/").text, "html.parser") #visit the page

#<a class="teaser__link" -> this is the identified pattern that leads to all articles
articleList = page.find_all("a", class_="teaser__link", href=True)

for article in articleList: #iterate over the list and print out the results
    print(article["href"])
```
