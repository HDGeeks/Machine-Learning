### Exercise 2: Writing to Files

Write the elements of *exampleList* to a file named *example.txt*. Each element should be written to a new line!


```python
exampleList = ["H", "I", "J", "K", "L"]
with open("example.txt", "w") as f: #open the file in write mode
    for item in exampleList: #iterate over the list and write each item to the file
        f.write(item + "\n") #write the item followed by a newline character to the file

```

### Exercise 3: Reading and Transforming Files

Read the contents of the *example.txt* file from the previous exercise, transform them to lowercase, and write the results to a new file.

```python
# Read the original file
with open("example.txt", "r") as f:
    lines = f.readlines() # read all lines from the file
with open("example2.txt", "w") as f: # open new the file in write mode
    for line in lines: # iterate over the lines and print them
        f.write(line.lower().strip() + "\n") # write the line in lowercase to the new file
```
