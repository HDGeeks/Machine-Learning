### Exercise 3: Dictionaries


```python
students = {
    "Anna": 19,
    "Carl": 21,
    "Anton": 34,
    "Bernd":20
}

# a) Add a new student
print('-' *20,"Aufgabe a:", '-'*50)

students['tim'] = 22  # Example of adding a new student
print(students)

# b) Calculate the average age
print('-' *20,"Aufgabe b:", '-'*50)

total_age = 0
for age in students.values():
    total_age += age  #Ist das gleiche wie total_age = total_age + age

average_age = total_age / len(students)
print(f"The average age is: {average_age}")


# c) (Nicely) print out every name and corresponding age
print('-' *20,"Aufgabe a:", '-'*50)

for name, age in students.items():
    print(f"Name: {name}, Age: {age}")

```

### Exercise 4: Shopping List

```python
# Create a program for a shopping list:
# 1. A user can insert items via the input() function
# 2. If the user inputs "done", insertion stops
# 3. After insertion, all elements should be printed

shopping_list = []
while True:
    item = input("Please enter an item for your shopping list (or type 'done' to finish): ")
    if item.lower() == 'done':
        break
    shopping_list.append(item)

    
print("Your shopping list contains:")
print(shopping_list)
```

### Exercise 5: More list operations


```python
grades = [1, 2, 1, 3, 2, 4, 1, 5]

# a) Calculate the average of the grades
avgerage =sum(grades)/len(grades)
print(avgerage)
# b) Determine the frequency of the number 1 in the list (count the occurences)
frequency_of_1 = grades.count(1)
print(frequency_of_1)

# c) Create a list that only contains grades greater than 3
new_list = []
for grade in grades:
    if grade > 3:
        new_list.append(grade)
print(new_list)
```
