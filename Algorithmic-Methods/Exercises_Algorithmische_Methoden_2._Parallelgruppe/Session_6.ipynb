# Implementierung von Datenstrukturen (Beispiel)

Implementieren Sie zwei Methoden, **enque** und **dequeue**, für die Datenstruktur **"Warteschlange"** (Queue). (https://de.wikipedia.org/wiki/Warteschlange_(Datenstruktur)). Verwenden Sie **zwei Python-Listen** als Stapel (Stacks) und implementieren Sie die Funktionen unter Verwendung der Methoden **append()** und **pop()**.

````python
class Queue:

    def __init__(self):
        self.stack1 = [] 
        self.stack2 = []

    def enqueue(self,val):
        while len(self.stack2) !=0: 
            self.stack1.append(self.stack2.pop()) #Pop ist eine Methode der Klassse Liste und gibt per default den letzten wert der liste aus
        self.stack1.append(val)

    def dequeue(self):
        while len(self.stack1) != 0:
            self.stack2.append(self.stack1.pop())
        return self.stack2.pop()

q = Queue()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)

print(q.dequeue())
print(q.dequeue())
print(q.dequeue())
````

Erweitern Sie die unten stehende einfache verkettete Liste zu einer doppelt verketteten Liste (https://en.wikipedia.org/wiki/Linked_list).

Die Kommentare im Code sind ein Ausgangspunkt und deuten an, wo Sie etwas hinzufügen müssen.

````python
class ListElement():
  def __init__(self,value):
    self.value = value
    self.next = None
    # HIER ERGÄNZEN: Füge einen Zeiger auf das vorherige Element hinzu.
    self.prev = None
    

class SimpleLinkedList():
  def __init__(self):
    self.head = None
    # HIER ERGÄNZEN: Speichere zusätzlich einen Zeiger auf das letzte Element.
    self.tail = None

  def addElement(self,value):
    # HIER ERGÄNZEN: Implementiere die Logik zum Hinzufügen eines Elements.
    # Achte darauf, 'head', 'tail', 'next' und 'previous' korrekt zu setzen.
    new_element = ListElement(value)
    if self.head == None:
      self.head = new_element
      self.tail = new_element # only element is head and tail
    else:
      self.tail.next = new_element # set next of tail to new element
      new_element.prev = self.tail # set previous of new element to tail
      self.tail=new_element # update tail to new element
      # HIER ERGÄNZEN: Setze den 'previous' Zeiger, 'next' Zeiger und 'tail'

 
  # Gibt alle Elemente der Liste in der Reihenfolge von 'head' nach 'tail' aus.
  def printList(self):
    current = self.head
    print("Inhalt der Liste (Vorwärts):")
    # Durchlaufe die Liste, solange 'current' nicht 'None' (das Ende) ist.
    while current is not None:
      print(current.value)
      # Gehe zum nächsten Element in der Kette.
      current = current.next

  # Gibt alle Elemente der Liste in umgekehrter Reihenfolge aus (von 'tail' nach 'head').
  def printListReverse(self):
    current = self.tail
    print("\nInhalt der Liste (Rückwärts):")
    # Wir starten am 'tail' und gehen zurück, solange 'current' nicht 'None' ist.
    while current is not None:
      print(current.value)
      # Gehe zum vorherigen Element mithilfe des 'previous'-Zeigers.
      current = current.previous

  
sll= SimpleLinkedList()
sll.addElement(1)
sll.addElement(2)
sll.addElement(3)
sll.addElement(4)
sll.addElement(5)
sll.addElement(6)

sll.printList()
````

# Inheritance (Example)



````python
class Instrument:
    """Basisklasse für alle Musikinstrumente."""
    def __init__(self, name, material):
      self.name = name
      self.material = material

    def play_sound(self):

      print(f"Das {self.name} aus {self.material} macht einen Ton.")

    def print_info(self):
      print(f"--- {self.name} | Material: {self.material} ---")

class StringInstrument(Instrument):
    """Erbt von Instrument und ist die Basis für Saiteninstrumente."""
    def __init__(self, name, material, num_strings):
      super().__init__(name, material)
      self.num_strings = num_strings

    def tune(self):
      print(f"Die {self.num_strings} Saiten der {self.name} werden gestimmt.")

class Guitar(StringInstrument):
    """Spezialisiertes Saiteninstrument."""
    def __init__(self, name, material, num_strings, is_electric):
      super().__init__(name, material, num_strings)
      self.is_electric = is_electric

    def play_sound(self):
      """Überschreibt play_sound mit spezifischem Verhalten."""
      action = "gezupft" if not self.is_electric else "elektrisch verzerrt"
      print(f"Die Gitarre erzeugt einen Klang: {action}.")

class Drum(Instrument):
    """Ein Schlaginstrument, erbt direkt von Instrument."""
    def __init__(self, name, material, diameter):
      super().__init__(name, material)
      self.diameter = diameter

    def play_sound(self):
      """Überschreibt play_sound mit spezifischem Verhalten."""
      print(f"Die Trommel (Ø{self.diameter}cm) erzeugt einen Klang durch Schlagen des Fells.")

# 1. Objekte erstellen
guitar = Guitar("Konzertgitarre", "Holz", 6, False)
drum = Drum("Snare-Drum", "Metall/Fell", 35)

# 2. Polymorphie demonstrieren (Aufruf derselben Methode)
instruments = [guitar, drum]

for instrument in instruments:
    instrument.print_info()
    # Der Aufruf ist IMMER .play_sound(), aber das Verhalten ändert sich
    instrument.play_sound()
    
# 3. Spezialisierte Methode demonstrieren
print("\n--- Spezialisierte Aktion ---")
guitar.tune()
# drum.tune() würde einen Fehler erzeugen, da Drum diese Methode nicht erbt/besitzt
````

# Implementing Data Structures (Example)
**Exercise (and Homework): Media Classes**

1. Create a base class **Media** (or **Medium**) that contains the following:

* Basic attributes such as **Title**, **Year**, **Creator**, **Genre**

* Methods for retrieving the attributes

2. Create specialized subclasses that additionally contain specific information:

* **Movie** (add: **Duration**, **Director**, **Studio**)

* **TV Series** (add: **Number of Seasons/Episodes**, **Production Period**)

* **Podcast** (add: **Duration**, **Frequency**)

* **Article** (add: **Journal/Magazine**, **DOI**, **Keywords**)

````python
class Media:
    def __init__(self, title, year,creator,genre):
        self.title = title
        self.year = year
        self.creator = creator
        self.genre = genre

    def get_info(self):
        return f"{self.title} ({self.year}) - Creator: {self.creator}, Genre: {self.genre}"
    
class TVShow(Media):
    def __init__(self, title, year, creator, genre,number_of_seasons,production_year):
        super().__init__(title, year, creator, genre)
        self.seasons = number_of_seasons
        self.production_year = production_year

    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Seasons: {self.seasons}, Production Year: {self.production_year}"
    
class Podcast(Media):
    def __init__(self, title, year, creator, genre,number_of_episodes,production_year):
        super().__init__(title, year, creator, genre)
        self.episodes = number_of_episodes
        self.production_year = production_year

    def get_info(self):
        base_info = super().get_info()
        return f"{base_info}, Episodes: {self.episodes}, Production Year: {self.production_year}"
````
