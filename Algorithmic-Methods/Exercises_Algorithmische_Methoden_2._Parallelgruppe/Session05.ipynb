## Rekursion vs. Iteration

Vergange Woche haben wir die Fibonacci-Folge iterativ implementiert.

Diese Woche möchten wir das als Rekursive Funktion machen. 

Im Kontext der Programmierung (z. B. in Python) bedeutet rekursiv, dass eine Funktion sich selbst aufruft, um ein Problem in kleinere Teilprobleme zu zerlegen. Dies geschieht so lange, bis eine sogenannte Abbruchbedingung erreicht wird, die verhindert, dass die Rekursion unendlich weiterläuft.

**Aufbau einer rekursiven Funktion**
- Basisfall (Abbruchbedingung): Der einfachste Fall, bei dem keine weitere Rekursion nötig ist.

- Rekursiver Fall: Die Funktion ruft sich selbst mit einem vereinfachten oder kleineren Eingabewert auf.
