## Task 2: List Comprehension


```python
#You are given a list with words:
words = ["Python", "Programmierung", "Liste", "Comprehension", "Lambda", "Map"]

# a) 
# Create a new list using a list comprehension. The new list should only contain words with more than 6 characters
more_than_6 = [word for word in words if len(word) > 6]
print(more_than_6)

#klassischer loop 
neue_liste=[]
for word in words:
    if len(word)>6:
        neue_liste.append(word)
print(neue_liste)

#list comprehension
neue_liste=[word for word in words if len(word)>6]
print(neue_liste)


# b)
# Erstellen Sie eine weitere Liste, die die Länge jedes Wortes aus der Originalliste enthält.

laengen_liste = [len(word) for word in words]
print(laengen_liste)

```

## Task 4: Regular Expressions


```python
# Given is a list of urls:
urls = [
    "https://www.examplecom",
    "http://subdomain.example.co.uk/path",
    "ftp://invalid.com",
    "https://no-tld",
    "not-a-url.com",
    "https://space domain.com",
    "https://example.com/path?param=value"
]

# b) Write a function to check if an URL is valid. Use regular expressions:
# A valid URL shoul have the following properties:

# Start with http:// or https://
# Should not contain any whitespace
# should have a dot followd by two or 3 characters 

import re
def is_valid_url(url):
    pattern=r"^(https|http)://[\S]+\.[a-zA-Z]{2,3}"
    if re.match(pattern, url):
        return True
    else: 
        return False


for url in urls:
    print(url)
    print(is_valid_url(url))
```

```python
# You are given a list of email adresses:
emails = [
    "test@example.com",
    "invalid.email@com",
    "user@domain.co.uk",
    "no@dots",
    "@nodomain.com",
    "spaces @domain.com",
    "valid.email+label@domain.com"
]

# a) Write a function that uses regular expressions to evaluate the validity of an email adress; adresses should:

# Contain the @ symbol
# Have at least one dot (.) after the @ symbol
# Not have any whitespaces

def is_valid_email(email):
    pattern=r"^[\S]+@[\S]+\.[a-zA-Z]{2,3}$"
    if re.match(pattern, email):
        return True
    else: 
        return False
    
for email in emails:
    print(email)
    print(is_valid_email(email))
```
