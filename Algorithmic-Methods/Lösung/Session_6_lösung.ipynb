# Klassen in Python 

````python
class Car:
    def __init__(self, Type, Brand, Speed):
            
        self.Type= Type
        self.Brand = Brand
        self.Speed =Speed

    def drive(self):
        print('brumm')
    
    def stop(self):
        print('skrrr')

A4=Car(Type='Kombi', Brand= 'Audi', Speed='200')


# Abfrage von Attributen ohne Klammern, da hier nichts "übergebn" wird
print(A4.Type)
print(A4.Brand)
print(A4.Speed)

# Aufrufen von Methoden mit Klammern, da wir den Parameter Self übergeben wollen. Dieser darf jedoch nicht ausgeschrieben werden
A4.drive()
A4.stop()
````

# Implementierung von Datenstrukturen (Beispiel)

Implementieren Sie zwei Methoden, **enque** und **dequeue**, für die Datenstruktur **"Warteschlange"** (Queue). (https://de.wikipedia.org/wiki/Warteschlange_(Datenstruktur)). Verwenden Sie **zwei Python-Listen** als Stapel (Stacks) und implementieren Sie die Funktionen unter Verwendung der Methoden **append()** und **pop()**.

````python
class Queue:

    def __init__(self):
        self.stack1 = [] 
        self.stack2 = []

    def enqueue(self,val):
        while len(self.stack2) !=0: 
            self.stack1.append(self.stack2.pop()) #Pop ist eine Methode der Klassse Liste und gibt per default den letzten wert der liste aus
        self.stack1.append(val)

    def dequeue(self):
        while len(self.stack1) != 0:
            self.stack2.append(self.stack1.pop())
        return self.stack2.pop()

q = Queue()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)

print(q.dequeue())
print(q.dequeue())
print(q.dequeue())
````

## **Aufgabe: 1**

Erweitern Sie die unten stehende einfache verkettete Liste zu einer doppelt verketteten Liste (https://en.wikipedia.org/wiki/Linked_list).

Die Kommentare im Code sind ein Ausgangspunkt und deuten an, wo Sie etwas hinzufügen müssen.

````python
class ListElement():

    # The constructor initializes a new list element.

    def __init__(self, value):

      # The value that this element stores (e.g., a number or a string).

      self.value = value

      # The 'next' pointer: It points to the *next* element in the list.

      # 'None' means that it is the last element in the chain.

      self.next = None

      # The 'previous' pointer: It points to the *previous* element in the list.

      # This makes the list a doubly linked list.

      self.previous = None


class SimpleLinkedList():

  # The constructor initializes an empty list.

  def __init__(self):

    # 'head': Always points to the *first* element of the list.

    # For an empty list, it is 'None'.

    self.head = None

    # 'tail' (end): Always points to the *last* element of the list.

    # Storing the end allows for quick addition.

    self.tail = None

    # Adds a new element to the end of the list.

def addElement(self, value):

  new_element = ListElement(value)

  # Case 1: The list is empty

  if self.head is None:

    # Since it is the only element, it is both the head and the end.

    self.head = new_element

    self.tail = new_element

  # Case 2: The list already contains elements

  else:

    # 1. We update the 'next' pointer of the *current* end.

    # It should now point to the new element.

    self.tail.next = new_element

    # 2. We set the 'previous' pointer of the *new* element.

    # It points back to the element that was previously the end.

    new_element.previous = self.tail

    # 3. We update the 'tail' pointer of the list itself.

    # The new element is now the new end of the list.

    self.tail = new_element

# Prints all elements of the list in order from 'head' to 'tail'.

def printList(self):

  current = self.head

  print("List contents (forward):")

# Iterate through the list while 'current' is not 'None' (the end).

  while current is not None:

    print(current.value)

  # Go to the next element in the chain.

    current = current.next

# Prints all elements of the list in reverse order (from 'tail' to 'head').

def printListReverse(self):

  current = self.tail

  print("\nList contents (backwards):")

  # We start at 'tail' and go back as long as 'current' is not 'None'.

  while current is not None:

    print(current.value)

  # Go to the previous element using the 'previous' pointer.

    current = current.previous

sll = SimpleLinkedList()
sll.addElement(1)
sll.addElement(2)
sll.addElement(3)
sll.addElement(4)
sll.addElement(5)
sll.addElement(6)

sll.printList()
sll.printListReverse()
````

# Inheritance (Example)



````python
class Instrument:
    """Basisklasse für alle Musikinstrumente."""
    def __init__(self, name, material):
      self.name = name
      self.material = material

    def play_sound(self):

      print(f"Das {self.name} aus {self.material} macht einen Ton.")

    def print_info(self):
      print(f"--- {self.name} | Material: {self.material} ---")

class StringInstrument(Instrument):
    """Erbt von Instrument und ist die Basis für Saiteninstrumente."""
    def __init__(self, name, material, num_strings):
      super().__init__(name, material)
      self.num_strings = num_strings

    def tune(self):
      print(f"Die {self.num_strings} Saiten der {self.name} werden gestimmt.")

class Guitar(StringInstrument):
    """Spezialisiertes Saiteninstrument."""
    def __init__(self, name, material, num_strings, is_electric):
      super().__init__(name, material, num_strings)
      self.is_electric = is_electric

    def play_sound(self):
      """Überschreibt play_sound mit spezifischem Verhalten."""
      action = "gezupft" if not self.is_electric else "elektrisch verzerrt"
      print(f"Die Gitarre erzeugt einen Klang: {action}.")

class Drum(Instrument):
    """Ein Schlaginstrument, erbt direkt von Instrument."""
    def __init__(self, name, material, diameter):
      super().__init__(name, material)
      self.diameter = diameter

    def play_sound(self):
      """Überschreibt play_sound mit spezifischem Verhalten."""
      print(f"Die Trommel (Ø{self.diameter}cm) erzeugt einen Klang durch Schlagen des Fells.")

# 1. Objekte erstellen
guitar = Guitar("Konzertgitarre", "Holz", 6, False)
drum = Drum("Snare-Drum", "Metall/Fell", 35)

# 2. Polymorphie demonstrieren (Aufruf derselben Methode)
instruments = [guitar, drum]

for instrument in instruments:
    instrument.print_info()
    # Der Aufruf ist IMMER .play_sound(), aber das Verhalten ändert sich
    instrument.play_sound()
    
# 3. Spezialisierte Methode demonstrieren
print("\n--- Spezialisierte Aktion ---")
guitar.tune()
# drum.tune() würde einen Fehler erzeugen, da Drum diese Methode nicht erbt/besitzt
````
