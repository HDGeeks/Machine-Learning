# Task I:



Use *requests* and *bs4* to gather links to all current articles on the Tagesschau.de news outlet. The links should be saved in a list (make sure to only save article links)

Step 1: Visit <a href=https://www.tagesschau.de/>this page</a> and think of how to find the articles in the htlm sourcecode.

Step 2: Have a look at the html source and identify the desired elements.

Step 3: Use the *find_all()* method to retrieve all relevant elements.

```python
import requests
import json #for later use
from bs4 import BeautifulSoup

page = BeautifulSoup(requests.get("https://www.tagesschau.de/").text, "html.parser") #visit the page

#<a class="teaser__link" -> this is the identified pattern that leads to all articles
articleList = page.find_all("a", class_="teaser__link", href=True)
print(type(articleList)) #check the type of the articleList variable
print(f'Found {len(articleList)} articles') #print the list of articles
for article in articleList: #iterate over the list and print out the results
    print(article["href"])
```

# Task II:

Visit all links and collect a corpus of the articles.

The first step is to visit one page and locate the article within the sourcecode:

```python
# https://www.tagesschau.de/inland/innenpolitik/verteidungsausgaben-debatte-trump-100.html => example article page
articlePage = BeautifulSoup(requests.get("https://www.tagesschau.de/inland/innenpolitik/verteidungsausgaben-debatte-trump-100.html").text,"html.parser")
article = articlePage.find("script", type="application/ld+json").text #identified pattern

print(type(article)) #check the type of the article variable
#print(article[:1000])

```

```python
print(article['articleBody'])
```

```python
articleJson = json.loads(article.replace("\n","")) #the article is stored in a json format and needs to be transformed from string to json
articleJson['articleBody'] #get the actual article
```

In the second step, the list of links can be used to access all articles:

```python
baseUrl = "https://www.tagesschau.de" #the base url; needs to be combined with the found hrefs
articles = [] #empty list to store articles
for article in articleList:
  if article["href"].startswith("/") and not article["href"].startswith("/multimedia"): #filter out unwanted elements that might cause problems
    articlePage = BeautifulSoup(requests.get(baseUrl + article["href"]).text,"html.parser") #gathering the data
    article = articlePage.find("script", type="application/ld+json").text
    articleJson = json.loads(article.replace("\n",""))
    articles.append(articleJson['articleBody'])

print (len(articles))
print (articles[1])
```

# Task 3

Find all links from a wikipedia article!

```python
# URL of the Wikipedia page
url = "https://de.wikipedia.org/wiki/Trier"

headers = {
    'User-Agent': 'MyLearningBot/1.0 (myemail@example.com)'
}

# Fetch the webpage content WITH the headers
response = requests.get(url, headers=headers)

# Fetch the webpage content
#response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
print(soup)
for link in soup.find_all('a', href=True):
  print (link)
```

```python
# Find all href links referring to Wikipedia itself
wikipedia_links = []
for link in soup.find_all('a', href=True):
    href = link['href']
    if href.startswith('/wiki/') and not "Datei:" in href:  # Wikipedia internal links
        wikipedia_links.append(f"https://de.wikipedia.org{href}")
```

```python
# Print the extracted links
for wiki_link in wikipedia_links[:10]:
    print(wiki_link)
```

Visit all links that refer to other articles and save the article texts!

```python
article_texts= []

for wiki_link in wikipedia_links[:10]:
    # Fetch the webpage content
    response = requests.get(wiki_link,  headers=headers)
    soup = BeautifulSoup(response.content, 'html.parser')
    content = soup.find('div', {'id': 'mw-content-text'})  # text der Seite
    print(content)
    paragraphs = content.find_all('p')  # Finde alle paragraphen tags

    # Find alle praragraphen in eienm Text und füge sie zu einem Text zusammen
    article_text = "\n".join([p.get_text(strip=True) for p in paragraphs])
    article_texts.append(article_text)

    print(article_text)
```
