## Aufgabe 3: Zip and Dictionaries


```python
# Two lists are provided:

students = ["Anna", "Ben", "Clara", "David"]
grades = [1.3, 2.7, 1.0, 2.3]

# a)
# Use zip(), to create a Dictionary that takes students as keys and grades as values 
dictonary = {}
for key, value in zip(students, grades):
    dictonary[key]=value


print(dictonary)
```

## Task 4: Regular Expressions


```python
# Given is a list of urls:
urls = [
    "https://www.examplecom",
    "http://subdomain.example.co.uk/path",
    "ftp://invalid.com",
    "https://no-tld",
    "not-a-url.com",
    "https://space domain.com",
    "https://example.com/path?param=value"
]

# b) Write a function to check if an URL is valid. Use regular expressions:
# A valid URL shoul have the following properties:

# Start with http:// or https://
# Should not contain any whitespace
# should have a dot followd by two or 3 characters 

import re

def validate_url(email):
    # Das RegEx-Muster
    # ^[^\s]+@[^\s]+\.[^\s]+$
    
    #pattern=r"^(https?|http)://" # ^ zeigt an das am Anfnag des Strings gesucht wird und ob es mit https oder https startet, gefolgt von "//"
    #pattern=r"^(https|http)://[\S]+\."  # prüft ob Leerzeichen zwischen "//" und dem ersten Punkt sind
    pattern=r"^(https|http)://[\S]+\.[a-zA-Z]{2,3}" # überprüft ob nach dem punkt 2 oder 3 Buchstaben sind
    
    # re.match() prüft, ob das Muster am Anfang des Strings übereinstimmt
    if re.match(pattern, email):
        return True
    else:
        return False

for url in urls:
    print(url)
    print(validate_url(url))
```

```python
# You are given a list of email adresses:
emails = [
    "test@example.com",
    "invalid.email@com",
    "user@domain.co.uk",
    "no@dots",
    "@nodomain.com",
    "spaces @domain.com",
    "valid.email+label@domain.com"
]

# a) Write a function that uses regular expressions to evaluate the validity of an email adress; adresses should:

# Contain the @ symbol
# Have at least one dot (.) after the @ symbol
# Not have any whitespaces
def validate_email(email):


 # Das RegEx-Muster:
    # ^                # Start des Strings
    # [\S]+            # Mindestens ein Nicht-Leerzeichen (Benutzername)
    # @                # Gefolgt vom @-Symbol
    # [\S]+            # Mindestens ein Nicht-Leerzeichen (Subdomain/Domain)
    # \.               # Gefolgt von einem literalischen Punkt (.)
    # [\S]+            # Gefolgt von mindestens einem Nicht-Leerzeichen (Top-Level-Domain/TLD)
    # $                # Ende des Strings.
    
    # Da [\S] (Nicht-Leerzeichen) überall verwendet wird, ist Regel 3 abgedeckt.
    # Da @ und \. (nach @) vorhanden sind, sind Regel 1 und 2 abgedeckt.
    pattern = r"^[\S]+@[\S]+\.[\S]+$"
    
    # re.match() prüft, ob das Muster am Anfang des Strings übereinstimmt.
    # Da wir ^ und $ verwenden, muss der gesamte String übereinstimmen.
    if re.match(pattern, email):
        return True
    else:
        return False

list(map(validate_email, emails))
```
