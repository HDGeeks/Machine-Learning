## Aufgabe 3: Zip and Dictionaries


```python
# Two lists are provided:

students = ["Anna", "Ben", "Clara", "David"]
grades = [1.3, 2.7, 1.0, 2.3]

# a)
# Use zip(), to create a Dictionary that takes students as keys and grades as values 
dictonary = {}
for key, value in zip(students, grades):
    dictonary[key]=value


print(dictonary)
```

## Task 4: Regular Expressions


```python
# Given is a list of URLs:
urls = [
    "https://www.examplecom",
    "http://subdomain.example.co.uk/path",
    "ftp://invalid.com",
    "https://no-tld",
    "not-a-url.com",
    "https://space domain.com",
    "https://example.com/path?param=value"
]

# b) Write a function to check if a URL is valid using regular expressions.
# A valid URL should have the following properties:
# - Start with http:// or https://
# - Should not contain any whitespace
# - Should have a dot followed by two or three letters (e.g., .com, .de)

import re

def validate_url(url):
    # The regular expression pattern
    
   
    
    # pattern = r"^(https?|http)://"
    # ^ indicates the beginning of the string.
    # This checks whether the string starts with http or https followed by "://".
    
    # pattern = r"^(https|http)://[\S]+\."
    # This checks whether there are no whitespace characters
    # between "://" and the first dot.
    
    pattern = r"^(https|http)://[\S]+\.[a-zA-Z]{2,3}"
    # This checks whether after the dot there are exactly
    # two or three alphabetic characters (e.g., .com, .de)
    
    # re.match() checks if the pattern matches at the beginning of the string
    if re.match(pattern, url):
        return True
    else:
        return False

for url in urls:
    print(url)
    print(validate_url(url))
```

```python
# You are given a list of email addresses:
emails = [
    "test@example.com",
    "invalid.email@com",
    "user@domain.co.uk",
    "no@dots",
    "@nodomain.com",
    "spaces @domain.com",
    "valid.email+label@domain.com"
]

# a) Write a function that uses regular expressions to evaluate
# whether an email address is valid.
# The addresses should:
# - Contain the @ symbol
# - Have at least one dot (.) after the @ symbol
# - Not contain any whitespace

import re

def validate_email(email):

    # The regular expression pattern:
    # ^                # Start of the string
    # [\S]+            # At least one non-whitespace character (username)
    # @                # Followed by the @ symbol
    # [\S]+            # At least one non-whitespace character (subdomain/domain)
    # \.               # Followed by a literal dot (.)
    # [\S]+            # Followed by at least one non-whitespace character (Top-Level-Domain)
    # $                # End of the string
    
    # Since [\S] (non-whitespace) is used everywhere,
    # rule 3 (no whitespace) is covered.
    # Since @ and \. (after @) are required,
    # rules 1 and 2 are covered.
    
    pattern = r"^[\S]+@[\S]+\.[\S]+$"
    pattern2 = r"^[\S]+@[\S]+\.([a-z]{2,3})$"
    
    # re.match() checks whether the pattern matches
    # at the beginning of the string.
    # Because we use ^ and $, the entire string must match.
    if re.match(pattern2, email):
        return True
    else:
        return False

list(map(validate_email, emails))
```
