```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data

np.random.seed(42)  # zur Reproduzierbarkeit
x = np.linspace(0, 10, 50)
y = np.sin(x) + np.random.normal(0, 0.1, 50)
```

```python
print(type(y))
print(y)
```

## Task 2: Add Statistics

Calculate and add to the plot:
- Minimum value line
- Maximum value line
- Average value line

Use horizontal lines (hint: plt.axhline) to show these statistics.

```python
# Your code here
plt.title('ein Graph')
plt.xlabel('x Werte')
plt.ylabel('y Werte')
plt.grid(True)
plt.plot(x, y, marker='o', linestyle='--', color='blue')


#minimum 
plt.axhline(min(y))
#maximum 
plt.axhline(max(y))
#avg 
plt.axhline(sum(y)/len(y))

```

## Task 3: Fill Between Min and Max

Create a shaded area between the minimum and maximum values.
Hint: Use plt.fill_between()

```python
# Your code here

# Your code here
plt.title('ein Graph')
plt.xlabel('x Werte')
plt.ylabel('y Werte')
plt.grid(True)
plt.plot(x, y, marker='o', linestyle='--', color='blue')


#minimum 
plt.axhline(min(y))
#maximum 
plt.axhline(max(y))
#avg 
plt.axhline(sum(y)/len(y))

plt.fill_between(x=x, y1=min(y), y2=max(y),color= 'gray', alpha=0.2)
```
