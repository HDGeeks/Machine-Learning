A recursive function is a function that calls itself within its definition to solve a problem by breaking it down into smaller, more manageable sub-problems. The key to recursive functions is having a base case to terminate the recursion. Without a base case, or if the base case is never reached, the function can lead to infinite recursion.<br><br>

Stack Overflow: Excessive recursion can consume all the stack space, leading to a stack overflow error. Performance Issues: Recursive functions can be less efficient than iterative counterparts due to the overhead of multiple function calls and returns

	•	Encapsulation – Bundling data and the methods that operate on that data within a class, and controlling access through well-defined interfaces. 
	•	Abstraction – Exposing only essential features of an object while hiding implementation details to reduce complexity.

Q5 .
Define the function count empty lines, which takes the name of a text file as an argument and prints all lines in this file (on the screen) that are not empty or consist only of whitespace characters. The function returns a message (string) as a value that indicates the number of all empty lines in the file.


```python
def count_empty_lines(file_path):
    """
    Count the number of empty lines in a text file.

    Args:
        file_path (str): Path to the text file.
    """
    count = 0
    with open(file_path, 'r') as file:
        for line in file:
            if line.strip() == "":
                count += 1
            else:
                print(line.strip(), end='')
    # Construct the return message
    return_message = f"The file {file_path} contains {count} empty lines."
    return return_message
```

```python
class SocialMediaPost():
    def  __init__(self, content: str, author: str):
        self.content = content
        self.author = author

    def show (self):
        print (f"{self.author} posted:")
        print(f"{self.content}")

# Solution
class TwitterMediaPost(SocialMediaPost):
    def __init__(self, content: str, author: str, hashtags: list = None, media: list = None):
        """
        Initialize a SocialMediaPost object.

        Args:
            content (str): The content of the post.
            author (str): The author of the post.
            hashtags (list): A list of hashtags associated with the post.
            media (list): A list of media items (e.g., images, videos) associated with the post.
        """
        super().__init__(content, author)
        # Task 1 two extra attributes done 
        self.hashtags = hashtags if hashtags is not None else []
        self.media = media if media is not None else []
        
        # Task 2 is to perform content atribute restriction to 280 characters or any example limit
        if len(self.content) > 280:
            raise ValueError("Content exceeds 280 characters limit.")
    # Task 3 implement any class method
    def add_hashtag(self, hashtag: str):
        if not hashtag.startswith('#'):
            hashtag = '#' + hashtag  # Ensure hashtag starts with '#'.
        if len(self.content) + len(hashtag) + 1 <= 280:
            self.hashtags.append(hashtag)
        else:
            print("Cannot add hashtag. Character limit would be exceeded.")



    def show(self):
        """
        Display the post content along with author, hashtags, and media.
        """
        super().show()
        if self.hashtags:
            print("Hashtags:", ", ".join(self.hashtags))
        if self.media:
            print("Media items:", ", ".join(self.media))

    def __repr__(self):
        hashtags_str = " Hashtags: " + ", ".join(self.hashtags) if self.hashtags else ""
        media_str = " Media: " + ", ".join(self.media) if self.media else ""
        return f"Author: {self.author}\nContent: {self.content}{hashtags_str}{media_str}"
try:
    twitter_post = TwitterMediaPost(content="Check out my new blog post!", author="JohnDoe", hashtags=["#blog", "#writing"], media=["http://example.com/image.png"])
    print(twitter_post)
except ValueError as e:
    print(e)
```

Q9 . Define a function that takes any number of file names as an argument and processes the files with the spaCy library and outputs a report on the entities found for each document:

```python
import spacy

def entity_report(*files):
    nlp = spacy.load("en_core_web_sm")
    for f in files:
        text = open(f, "r", encoding="utf-8").read()
        doc = nlp(text)
        print(f)
        for ent in doc.ents:
            print(ent.text, ent.label_)
        print()
```
