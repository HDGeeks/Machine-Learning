# **Working with data**

```python
import requests
from bs4 import BeautifulSoup

page = BeautifulSoup(requests.get("https://www.tagesschau.de/").text, "html.parser") #visit the page

#<a class="teaser__link" -> this is the identified pattern that leads to all articles
articleList = page.find_all("a", class_="teaser__link", href=True)

baseUrl = "https://www.tagesschau.de" #the base url; needs to be combined with the found hrefs
articles = [] #empty list to store articles
for article in articleList[:1]:
  if article["href"].startswith("/") and not article["href"].startswith("/multimedia"): # filter out unwanted elements that might cause problems
    articlePage = BeautifulSoup(requests.get(baseUrl + article["href"]).text,"html.parser") #gathering the data
    article = articlePage.find("script", type="application/ld+json").text
    print (article)
    articles.append(article)

articles[0]
```

# **JSON**
The data above is formatted in json format. JSON stands for **J**ava**S**cript **O**bject **N**otation and is a lightweight data exchange format.

JSON is built on two structures:

- A collection of name/value pairs. In various languages, this is realized as an object, record, struct, dictionary, hash table, keyed list, or associative array.
- An ordered list of values. In most languages, this is realized as an array, vector, list, or sequence.

These are universal data structures. Virtually all modern programming languages support them in one form or another. It makes sense that a data format that is interchangeable with programming languages also be based on these structures. (https://www.json.org/json-en.html)

JSON can be easily processed by machines, while still remaining human-readable.

The built-in Python library *json* allows working with json data from a Python environment.
https://docs.python.org/3/library/json.html


```python
type(articles[0])
```

```python
import json

article = json.loads(articles[0].replace("\n", "")) #removing the newline characters and loading the string as json
article
```

According to the conversion table below, the json file is converted to their corresponding Python objects:

JSON |Python
---|---

string|str

number (real)|float
true|True

As indicated by its nature as a data exchange format, it is possible to save json objects as files. The convention is to use *.json* as file ending.

```python
#write dictionary to .json
with open("example.json", "w") as f:
  json.dump(exampleDict, f)
```

```python
!ls
```

```python
#read it back to from file
with open("example.json", "r") as f:
  print(json.load(f))
```

***Note the difference between *load()* and *loads()*!!!***

Both methods deserialize json objects to Python, but they expect a different input type:

- loads(): expects a string (or bytes, or bytearray)
- load(): expects a file object

The same applies to dump(s) accordingly; dump() serializes to a file object, which is used for writing json to files. dumps() serializes to a json formatted string Object.

# **File I/O**
Instead of just printing the output of a program, it is also possible to write the output of a program to a file for later use.
Within a Python program, a file is represented as a *file object*

File objects are created with the *open* function. By providing a file path to the *open()* function, a program is able to read and manipulate the underlying file.

https://docs.python.org/3/glossary.html#term-file-object
https://docs.python.org/3/tutorial/inputoutput.html#reading-and-writing-files

```python
#writing a string to a file
with open("example.txt", "w") as f:
  f.write("This is an example")
```

```python
!ls
```

```python
#reading a file to a string
with open("example.txt", "r") as f:
  print(f.read()) #reads the entire file
```

In the examples above, the *open()* function takes two arguments. The first is the path to the file (both relative and absolute paths are possible). The second one indicates the so-called *mode*.

The mode describes how the file can be handled. Below is an overview of the most common modes:

mode|description
---|---
r|read-only
w|write-only
a|append
r+|read and write

Note that opening a file in *w* mode will overwrite an existing file with the same name. If the file does not exist, it will be created.

Therefore, if the desired behavior is to add something to an existing file, the *append* mode needs to be used.

```python
with open("example.txt", "a") as f: #using append mode to add a new line to the end of the example.txt file
  f.write("\nThis is a new line")
```

```python
with open("example.txt", "r") as f:
  print(f.read())
```

```python
#altering the file contents by string manipulation
with open("example.txt","r+") as f:
  content = f.read()
  content = content.replace("example", "REPLACED")
  f.seek(0) #moving back to the beginning of the file in order to overwrite it with the altered content
  f.write(content)
```

```python
with open("example.txt", "r") as f:
  print(f.read())
```

Instead of processing the file contents all at once, it might be desirable to process them line by line. This reduces memory consumption as only smaller parts of a file are loaded in at once.

```python
with open("example.txt", "r") as f:
  for line in f:
    print(line) #reading the file line by line
```

```python
with open("example.txt", "r") as f:
  lines = f.readlines() #readlines reads the file line by line and returns a list with every list element being one line
  print(lines)
```

An important concept when dealing with file is character encoding:



> *Character encoding is the process of assigning numbers to graphical characters, especially the written characters of human language, allowing them to be stored, transmitted, and transformed using computers. The numerical values that make up a character encoding are known as code points and collectively comprise a code space, a code page, or character map.*

(https://en.wikipedia.org/wiki/Character_encoding)

Encodings differ in the way how they represent characters. Encodings that use more Bits can represent a larger set of characters. This becomes especially important if characters from non-english latin characters need to be displayed:


```python
with open("encoding.txt", "w") as f:
  f.write("This is an example\n")
  f.write("Dies sind Umlaute: ä ö ü Ä Ö Ü\n")
  f.write("Дорогие друзья!\n")
```

```python
with open("encoding.txt", "r") as f:
  print(f.read())
```

```python
with open("encoding.txt", "r",encoding="ASCII") as f:
  print(f.read())
```

```python
with open("encoding.txt", "w",encoding="ASCII",errors='ignore') as f:
  f.write("This is an example\n")
  f.write("Dies sind Umlaute: ä ö ü Ä Ö Ü\n")
  f.write("Дорогие друзья!\n")
```

```python
with open("encoding.txt", "r",encoding='cp1140') as f:
  print(f.read()) #worst case => no errors and loss of information
```

Leaving the *encoding* parameter to default will lead to using the current operating system's default character encoding.

# **PICKLE**

If, instead of just saving data, complete data structures need to be saved, Python Objects can be (de-)serialized using the *Pickle* module.
(https://docs.python.org/3/library/pickle.html)


json|pickle
---|---
text serialization|binary serialization
human-readable|only machine-readable
interoperable|python specific
only built-in types|custom classes
save|potentially unsafe

```python
class Person: #example class
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def say_hello(self):
    print(f"Hello, my name is {self.name} and I am {self.age} years old.")
```

```python
import pickle
#create an instance of person and save it to file
person = Person("John", 30)
with open("person.pickle", "wb") as f: #wb is used for writing binary
  pickle.dump(person, f)
```

```python
!ls
```

```python
#read pickle file
with open("person.pickle", "rb") as f:
  reloadPerson = pickle.load(f)

reloadPerson.say_hello()
```

```python
class Person: #new class definition
  def __init__(self, name, age, location):
    self.name = name
    self.age = age
    self.location = location

  def say_hello(self):
    print(f"Hello, my name is {self.name}.")

  def where_am_i(self):
    print (f"I am in {self.location}")
```

Note that a change in the class definition can lead to inconsistencies!

```python
with open("person.pickle", "rb") as f:
  reloadPerson = pickle.load(f)

reloadPerson.say_hello()
print(type(reloadPerson))

newPerson = Person("John", 30, "Vienna")
newPerson.where_am_i()
print(type(newPerson))
```

# **Writing a dataframe to a file**

Saving a dataframe to a file requires specifying a file name.
The default column separator is ",". The *sep* argument can be used to specify a different separator.

```python
data = {'name' : ["Jack","John","Jane"],'age' : [29,39,49]}
df3 = pd.DataFrame(data=data)
df3.to_csv("names.csv")
df3
```

Note how the first column (containing the index) will also be written to the file!

```python
!ls
```

```python
df4 = pd.read_csv("names.csv")
df4
```

Upon reading the file, a new index is created, leading to the result above!

To prevent this redundant indexing, there are two options:

- setting the *index* argument to false when writing to .csv

```python
data = {'name' : ["Jack","John","Jane"],'age' : [29,39,49]}
df3 = pd.DataFrame(data=data)
df3.to_csv("names.csv",index=False)
df3 = pd.read_csv("names.csv")
df3
```

- specifying an index column during reading

```python
data = {'name' : ["Jack","John","Jane"],'age' : [29,39,49]}
df3 = pd.DataFrame(data=data)
df3.to_csv("names.csv")
df3 = pd.read_csv("names.csv",index_col=0)
df3
```

**It is important that the separator remains consistent across writing and reading the file:**

```python
data = {'name' : ["Jack","John","Jane"],'age' : [29,39,49],'biography':['Hi, my name is ...',"I am an ...","This is ..."]}
df3 = pd.DataFrame(data=data)
df3.to_csv("names.csv",sep=";")
df3 = pd.read_csv("names.csv",index_col=0) #sep defaults to , -> trying to parse the comma in "Hi, my ..." as separator!
df3
```

**Sampling**

Sampling describes the process of (randomly) drawing samples out of a Dataframe.

The *sample* method allows for both drawing an amount of samples (specified by an integer) or a fraction of the dataset as samples (specified by a float).

In the example of sampling it becomes apparent why the index is important. Without an index, it would not always be possible to keep track of which rows were drawn.

```python
df3 = pd.read_csv("names.csv",sep=";",index_col=0)
integerSample = df3.sample(n=2)
integerSample
```

```python
fractionSample = df3.sample(frac=0.33)
fractionSample
```

It is not possible to draw a fraction of a column as sample. Therefore, the amount of samples drawn is rounded up or down (depending of which is closer)

```python
fractionSample = df3.sample(frac=0.49)
fractionSample
```

```python
fractionSample = df3.sample(frac=0.5)
fractionSample
```
