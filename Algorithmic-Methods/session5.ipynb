# **Additional Keywords for loops**

There are two more important control flow tools that can be used in loops: **break** and **continue**

**Break** is used to break out of a while or a for loop. Once the keyword *break* is reached inside the loop, no more code inside the loop will be executed.

**Continue** is used to skip an interation of a loop. Once the *continue* keyword is reached, all subsequent code for the current iteration is skipped and the loop continues with the next iteration.



**Break Example:**

Below is an example situation where the the program awaits input and keeps on running until the correct input is inserted.

Note that the same behaviour could also be achieved by setting a boolean variable as condition for the while loop and change its value once the expected input is inserted.

```python
password = "1234"

while True:
  passwordInput = input("Enter your password:")
  if passwordInput == password:
    break
  print ("Try again!")

print ("Welcome")
```

**Continue Example**

An example usage for a continue statement would be a program where you are required to check conditions first before executing the actual code.

Using continue instead of if else blocks can improve code readability by reducing the required level of nesting.


```python
files = ["img1.jpg","img2.jpg","img3.jpg","img4.jpg","img5.jpg","img6.png","img7.jpg"]
outputFiles = []

for file in files:
  if file.endswith(".png"):
    continue
  outputFiles.append(file.upper())

print (outputFiles)
```

**Methods**

Similar to functions, methods also perform tasks, may have arguments and return types. They are different from functions only in the aspect that they are defined within a class or a data type. As an example, objects that belong to the string type can call the *lower()* method that returns a copy of the string with all characters in lowercase:

```python
exampleString = "Hi, this is a test!"
exampleString.lower()
```

Note the difference between invoking a method and a function! Methods always require a corresponding object that implements them. Functions exist on their own and can be called independently. If you try to use a method on a data type that does not implement said method, you will raise an error:

```python
exampleNumber = 15
exampleNumber.lower()
```

Methods can also have arguments. Take the *split()* method as an example, which takes a string as argument to determine on which character the original string should be split:

```python
exampleString.split(',')
```

Multiple methods can be chained together. They are evaluated from left to right:

```python
exampleString.lower().split()
```

The code above first executes the *lower()* method and subsequently uses its returned object to perform the *split()* method.

It is also possible to combine the expression above with a function call.

```python
len(exampleString.lower().split())
```

Here, we have used the len() function to determine the number of elements in the list that is returned by the split() method.

Note that methods can have different return types, even when they are implemented by the same type:

    string.lower() -> string
    string.split() -> list

**Defining functions**

Although it is possible to get quite far with using only built-in functions, types and their corresponding methods, it is often required to define a new function specific to a certain use-case. Functions are defined by using the **def** keyword, followed by a function name and parantheses that can contain the function argument(s):

    def doNothing():
      pass

The example above is a minimal function definition. It does not take any arguments and - by using the **pass** keyword - does nothing. In many cases, you want to define a return value for a function. This is achieved by using the **return** keyword:

    def doSomething():
      return "Something"

Again, the example above does not do anything particularly useful. It only returns the string "Something" as return value.
Let us assume that you repeatedly need to lower a string a then split it into a list for some task. You could define a function *lowerAndSplit* that takes a string as an argument, performs the *lower()* and the *split()* method and returns the new string:

```python
def lowerAndSplit(argumentString):
  lowered = argumentString.lower()
  splitToList = lowered.split()
  return splitToList
```

```python
testString = "Hi, this is an IMPORTANT TEST!"
lowerAndSplit(testString)
```

**Example - Textprocessing**

Consider the following wikipedia article from a few weeks ago. We have implemented some pre-processing steps and functionality to create a word count for this article:

```python
wikipediaArticle = "Risby is the site of a deserted village and former stately home in the civil parish of Rowley in the East Riding of Yorkshire, England. \
It is approximately 3.5 miles (5.6 km) south-west of Beverley and 1 mile (1.6 km) west of the A164 road. The area has an open partially wooded parkland, \
once one of the largest deer parks in Yorkshire, that is popular with local walkers. It also includes Risby Park, a farm, the Folly Lake Cafe. \
There are several fishponds in the ornamental lakes of the former hall, which have been operated as coarse fishing locations since 1990. \
The site also has an octagonal brick folly that is designated Grade II."
```

With our newly gained knowledge, we are now able to define these multi-step approaches as functions that will work on any provided string.

```python
def preProcess(article):
  preprocessedArticle = article.lower().replace('.','').replace(',','').replace('"','').replace('(','').replace(')','') #chaining all functions together
  preprocessedWordList = preprocessedArticle.split()
  return preprocessedWordList

def wordCount(wordList):
  countDictionary = {} #init empty dictionary; this will be used to store the words and their corresponding occurences
  for word in wordList: #iterate over the wordlist
    if word not in countDictionary: #if the current word is previously unseen, we create a new entry in the dictionary with the word as key and the number 1 as value
      countDictionary[word] = 1
    else: #if an entry with the current word as key already exists, we increase the value by 1
      countDictionary[word] += 1
  return countDictionary

def processAndCount(article):
  return wordCount(preProcess(article)) #chaining the functions together
```

```python
countDictionary = processAndCount(wikipediaArticle)
```

```python
def findMostFrequent(countDict):
  maxCount = 0
  maxWord = ""
  for key in countDict:
    if countDict[key] > maxCount:
      maxCount = countDict[key]
      maxWord = key
  return maxWord, maxCount
```

```python
findMostFrequent(countDictionary)
```

```python
randomArticle2 = "Piers Jourdain Bishop (born Hammersmith, London, 1956) is a British artist, and former child star. He is the eldest of three children. Within the 1970s, Bishop was a DJ on the north Norfolk circuit, and later in East London. He has been involved in starting up a number of businesses for arts and regeneration projects within the Lea Valley, and the North-West, including Artzone, and the Claverings company LTD. Currently, Piers is a practising artist, focussing mainly in painting and digital works, based on scientific themes. He lives in Morecambe, Lancashire, following leaving the Claverings company in 2002."
findMostFrequent(processAndCount(randomArticle2))
```
