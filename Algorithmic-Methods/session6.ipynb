# **Classes - Overview**

Python is an object-oriented programming language. Defining classes allows us to define custom objects or data structures.

**Instances and Classes**

All variables in Python belong to a certain data type, they are **instances** of said type. The different data types like integers, floats, etc. are implemented as classes. The class of a variable can be accessed by its \__class__ attribute:

```python
exampleString = "Hi there!"
exampleString.__class__
```

Note that the same information can be acquired by using the already known *type()* function:

```python
type(exampleString)
```

**Class methods**

A class defines which data can be stored in its instances and defines operations that can be performed on its instances. The *String* class for example stores text as a sequence of characters and provides a variety of methods that can be used on all instances of the String class.

```python
exampleString.lower()
```

**Class hierarchy**

Classes can form hierarchies, where subclasses inherit properties of its parent class(es). This can be imagined as forming a taxonomy, where the more general class (for example: vehicles) defines properties that all its subclasses have. Subclasses of *vehicle* could be *car* and *motorcycle*, that could define its specific properties.

# Defining classes

Classes are defined by using the **class** keyword. As per convention, class names start with a capital letter. The example below shows an absolute minimal class definition in Python:

```python
class ExampleClass():
  pass
```

This class does not define any data to be stored, nor any methods. However, by defining it, it becomes instantiable, which means that instances of this class can be created and assigned variable names:

```python
test1 = ExampleClass()

print (test1.__class__)

print (type(test1))
```

For a class to become actually useful, some functionality needs to be implemented. This can be achieved by defining variables within the class:

```python
class UsefulNumbers():
  pi = 3.141
  e = 2.718
```

In the example above, we have defined the *UsefulNumbers* class and defined two **class variables** within it. The number Pi and the number e. Class Variables can be accessed by all instances of said class:

```python
a = UsefulNumbers()
a.pi
```

```python
a.e
```

```python
b = UsefulNumbers()
b.pi
```

In addition to class variables, which are set for all instances of a class, there are also **instance variables**, that are set for each instance individually. Most commonly, they are defined within the special *__init__* method:

```python
class Student():
  is_human = True #class variable - a property that exists for all instances of a class

  def __init__(self,name): #self refers to the instance to be created
    self.name = name #instance variable
```

Defining the instance variables in the init method and adding them as arguments allows for setting this variable at instantiation:

```python
student1 = Student('John') #init student with name John
student2 = Student('Jane') #init student with name Jane

print (student1.name)
print (student1.is_human)
```

Note that the first argument for the init method is *self*, which is a reference to the object itself. All methods within a class need to have self as the first argument:

```python
class Student():
  is_human = True #class variable - a property that exists for all instances of a class

  def __init__(self,name): #self refers to the instance to be created
    self.name = name #instance variable

  def sayHello(): #Without referencing to itself, we get an error message
    print("Hello")

student3 = Student('Bob')
student3.sayHello() #by calling the method, we implicitly pass self as argument. The error is a result of the definition of the method as having zero parameters
```

```python
class Student():
  is_human = True #class variable - a property that exists for all instances of a class

  def __init__(self,name): #self refers to the instance to be created
    self.name = name #instance variable

  def sayHello(self): #Without referencing to itself, we get an error message
    print(f"Hello, my name is {self.name}")

student3 = Student('Bob')
student3.sayHello() #by calling the method, we implicitly pass self as argument. The error is a result of the definition of the method as having zero parameters
```

As already hinted, classes can be organised in hierarchies, with subclasses that can inherit properties (and methods) from their parent classes.

Consider the following example, where people associated with a university are modelled as classes. There are different types of people, as indicated by their roles, and different properties that need to be modelled.

Both students and employees are persons, so they both have a name and an age.

Students have a programme or subject that they are studying, whereas Employees belong to a certain department.

```python
class Person():
  def __init__(self,name,age): #properties that are shared across all subclasses
    self.name = name
    self.age = age

  def greetings(self): #method that will be available to all subclasses
    print (f"Hi, my name is {self.name} and I am {self.age} years old")

class Student(Person): #defining the class to inherit from
  def __init__(self,semester,programme,name,age):
    super().__init__(name,age) #calling the superclass init
    self.semester = semester
    self.programme = programme

class Employee(Person):
  def __init__(self,department,title,name,age):
    super().__init__(name,age)
    self.department = department
    self.title = title

professor = Employee("High speed computation", "Prof. Dr. Dr.","Jake", 91)
professor.greetings()
professor.name

student = Student("Physics",42,"HereTooLong",76)
student.greetings()
```

# Special Class Methods
Some class methods like \_\_init\_\_() and \_\_str\_\_() are special in the way that they are called. Unlike general class methods, those methods are not referred to by their name, but instead determine the behavior for other notations.

**\_\_init\_\_()** is a constructor that determines how a new class instance should be created. Consider the example below:

```python
class Showcase():
  def __init__(self,value):
    self.value = value
```

The init method is used to create an instance of the class *Showcase*, and expects a *value* to be passed as an argument. Unlike other methods, the init method is not called via its name (as it was the case with the *.lower()* method. Instead it is called via the Class Name:

    example = Showcase(5)

```python
example = Showcase(5)
print (type(example))
```

Similarly, the \_\_str\_\_() method is not called by explicitly referring to it via its name, but instead defines how the class instance should be represented when it is passed to the print() function. Not that the return type of this method needs to be a string!

```python
class Showcase2():
  def __init__(self,value):
    self.value = value

  def __str__(self):
    return self.value
```

```python
example2 = Showcase2(15) #Not a string!
print (example2)
```

```python
class Showcase2():
  def __init__(self,value):
    self.value = value

  def __str__(self):
    return str(self.value)

example2 = Showcase2(15) #Not a string!
print (example2)
```

Similarly, the \_\_len\_\_() method defines the behavior when the instance is passed as argument to the *len()* function. The return type is required to be an integer >= 0, otherwise an error will be thrown.

```python
class Showcase3():
  def __init__(self,value):
    self.value = value

  def __str__(self):
    return str(self.value)

  def __len__(self):
    return len(self.value)

example3 = Showcase3("Hi there")
len(example3)
```

# Linked List
An easy example of such a data structure is a simple linked list:

<img src=https://upload.wikimedia.org/wikipedia/commons/6/6d/Singly-linked-list.svg> </img>

A linked list can be used to store data. Every Element in the list contains some value and a pointer to the next element. The List itselfs stores only the first (head) element.

Common operations on lists are:
- adding an element to the (end of the) list
- traversing all elements
- checking whether an element is contained in the list
- determining the number of elements in the list

**List Element**
- needs to store the value
- needs to point to the next element

```python
class ListElement(): #A list element saves a value and its successor
  def __init__(self,value):
    self.value = value
    self.next = None

element = ListElement(15)
print (element)
```

**List Object**

- needs to store the first element (or head element)
- needs to implement common list functionalities

```python
class SimpleLinkedList(): #The list itself only saves the first element.
  def __init__(self):
    self.head = None

  def addElement(self,value): #adding an element
    if self.head == None: #if the list is empty, we set the new element to be the first element
      self.head = ListElement(value)
    else: #if not, we need to iterate to the last element, and the new element to be the next element
      current = self.head
      while current.next != None:
        current = current.next
      current.next = ListElement(value)

  def __str__(self):
    current = self.head
    strRep = "["
    while current != None:
      strRep += f" {current.value} "
      current = current.next
    strRep += "]"
    return strRep

  def __len__(self):
    length = 0
    current = self.head
    while current != None:
      length += 1
      current = current.next
    return length

newList = SimpleLinkedList()
newList.addElement(1)
newList.addElement(2)
newList.addElement(3)
print (newList)
print (len(newList))
```

# Stack
A stack is another data structure for storing elements. The common property of stacks is that only the last element that was added to the stack can be retrieved (Last in, first out):

<img src=https://upload.wikimedia.org/wikipedia/commons/e/e4/Lifo_stack.svg>

Stacks commonly imlement two methods:
- push: for adding elements to the top of the stack
- pop: for retrieving (and removing) the element at the top of the stack

A stack can be easily implemented in Python, as basic lists already implement this behavior with the *append()* and *pop()* methods.

```python
class Stack():
  def __init__(self):
    self.stack = []

  def push(self,element):
    self.stack.append(element)

  def pop(self):
    self.stack.pop()

  def __str__(self):
    return str(self.stack)

  def __len__(self):
    return len(self.stack)
```

```python
newStack = Stack()
newStack.push(5)
newStack.push(6)
newStack.push(8)
print (newStack)
newStack.pop()
print (newStack)
print (len(newStack))
```

The goal is to build a class that stores the corpus and provides functionalities to determine the relationship between word lenght and word frequency.

```python
class TextAnalyser():
  def __init__(self,corpus):
    self.corpus = corpus #for storing the original corpus
    self.cleanedCorpus = [] #for storing the pre-processed corpus
    self.frequencies = {} #for storing words (keys) and their frequencies (values)
    self.lenghts = {} #for storing words (keys) and their lengths in characters (values)
    self.avgLengthDict = {} #frequencies as keys, list of lenght as values

  def preprocess(self):
    for text in self.corpus:
      self.cleanedCorpus.append(text.lower())

  def determineFrequenciesAndLenghts(self):
    self.preprocess()
    for text in self.cleanedCorpus: #iterate over the corpus
      words = text.split()
      for word in words: #iterate over the words
        if word in self.frequencies:
          self.frequencies[word] += 1 #counting the word occurences
        else:
          self.frequencies[word] = 1 #initializing the word counter
          self.lenghts[word] = len(word) #determining the word length

  def analyse(self):
    self.determineFrequenciesAndLenghts()
    lengths = list(self.lenghts.values())
    frequencies = list(self.frequencies.values())

    for length,frequency in zip(lengths,frequencies):
      if frequency in self.avgLengthDict.keys():
        self.avgLengthDict[frequency].append(length) #append lenght to list to lenghts for given key
      else:
        self.avgLengthDict[frequency] = [length] #create key (frequency) value (list of lengths) pair

    for key in self.avgLengthDict:
      self.avgLengthDict[key] = sum(self.avgLengthDict[key])/len(self.avgLengthDict[key]) #calculate the average

  def __str__(self):
    representation = "Frequency\tAVG. Length\n"
    for key, value in sorted(self.avgLengthDict.items(), key=lambda x: x[0]):
      representation += f"{key}\t\t{value}\n"
    return representation

analyser = TextAnalyser(corpus)
analyser.analyse()
print (analyser)
```

The relation shown here is known as [Menzerath's law](https://en.wikipedia.org/wiki/Menzerath%27s_law), a linguistic law that can be phrased as:

"The larger a linguistic construct, the smaller its constituents"
