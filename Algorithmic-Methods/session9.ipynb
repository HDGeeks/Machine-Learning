**Importing**

"*Python code in one module gains access to the code in another module by the process of importing it. The import statement is the most common way of invoking the import machinery[...]*"

By using the *import* keyword, modules can be imported such that the corresponding functions, classes and their methods can be used.

The styleguide for imports can be found here: https://peps.python.org/pep-0008/#imports

The *howdy* module from above can be imported in various different ways that all come with their own specific behaviour:

```python
import howdy #standard import statement
howdy.howdy()
```

```python
import howdy as hy #using an alias that can be then used to access the module
hy.howdy()
```

```python
from howdy import howdy #import all contents of howdy to the current namespace
howdy()
```

It is also possible to import only certain functions from a module:

```python
from howdy import howdy
howdy()
```

```python
from howdy import howdy as hy
hy()
```

Imports from larger codebases can be specified to only import certain parts of the module:

```python
import torch.nn as nn #only imports the nn submodule from pytorch

nn.Linear(2,2)
```

The preferred way of importing modules are:

    import MODULE (as alias)

    from MODULE import FUNCTION/CLASS (as alias)

This makes it so that the imported functionality can be accessed by using the *module* namespace:

    MODULE.FUNCTION()

When using the

    from MODULE import *

syntax, all of the module's content will be available in the current namespace. This can have disadvantages when multiple modules implement a function of the same name:

```python
from goodbye import *
from howdy import *

howdy() #howdy() function from howdy
```

```python
from howdy import *
from goodbye import *

howdy() #howdy() function from goodbye
```

If instead the preferred import syntax is used, both functions will be available under their respective namespaces:

```python
import howdy
import goodbye

howdy.howdy()
goodbye.howdy()
```

```python
import goodbye
import howdy #the import order does not affect code behaviour anymore, because it is explicit which function needs to be called

howdy.howdy()
goodbye.howdy()
```

# **Local Modules**

The example modules *goodbye* and *howdy* from above are so-called local modules. Like every module, these can range from complex programs to simple scripts. One of the usecase would be to organize code in a local workspace, for instance moving utility functions to their own script in order to keep the code for the main functionality clean.

When moving functionality to a new script, it is important to consider that modules can not only contain class and function definitions, but also simple python expressions (like print statements):

```python
def xTimesTwenty(x): #defining a function
  return 20*x

print (xTimesTwenty(5)) #'testing' the function
```

The code above is the same as the code included in the *bad* module that will be imported below:

```python
import bad #import

print (bad.xTimesTwenty(2)) #calling the function
```

Note that there will be two outputs. One from the print statement from within the imported script (that will be executed) and one from current print statement. To prevent this from happening, all code that should not be executed on import should be put into a specific block of code:

```python
def xTimesTwenty(x): #defining a function
    return 20*x

if __name__ == "__main__":
    print (xTimesTwenty(5)) #'testing' the function
```

```python
import good

print (good.xTimesTwenty(2)) #calling the function
```

It is good practice to put all statements that are not imports, global variables, class definitions, or function definitions under

    if __name__ == "__main__":

as this will enable your scripts to be reusable without any side effects.

**spaCy**

spaCy provides lot of NLP functionalities, for instance methods for proper tokenization:

```python
import spacy
nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying U.K. startup for $1 billion")
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

Coming back to the word counting example from earlier in the lecture, we are now able to properly tokenize the texts and immediately count the words:

```python
randomWikipediaArticle = "Risby is the site of a deserted village and former stately home in the civil parish of Rowley in the East Riding of Yorkshire, England. \
It is approximately 3.5 miles (5.6 km) south-west of Beverley and 1 mile (1.6 km) west of the A164 road. The area has an open partially wooded parkland, \
once one of the largest deer parks in Yorkshire, that is popular with local walkers. It also includes Risby Park, a farm, the Folly Lake Cafe. \
There are several fishponds in the ornamental lakes of the former hall, which have been operated as coarse fishing locations since 1990. \
The site also has an octagonal brick folly that is designated Grade II."

doc = nlp(randomWikipediaArticle)
tokenList = []

for token in doc:
  tokenList.append(token.text.lower())

print (tokenList)
```

```python
wordDf = pd.DataFrame(tokenList, columns = ['Words'])
print (wordDf['Words'].value_counts())
```

In many cases, the printed error message provides an initial orientation (if not already the solution) for a way of fixing the issue.
