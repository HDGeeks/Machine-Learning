## 1. Bernoulli Distribution — MLE  
Random variable:  
- 1 = success  
- 0 = failure  

Observed data: a sequence like `[1,0,1,1,0,...]`.

Goal:  
Estimate θ = P(success) that best explains the data.

MLE:  
\[
\theta^\* = \frac{\sum_i x_i}{N}
\]

This is the *frequency of successes*.

```python
import numpy as np

# Simulated data
data = np.array([1,1,0,1,0,1,1])
theta_mle = data.mean()

theta_mle
```

## 2. Categorical Distribution — MLE

If X takes values in {A,B,C,...},  
MLE is simply the *relative frequency*:

\[
\theta_k^\* = \frac{\text{count of category k}}{N}
\]
